## Introducción al Encapsulamiento y acceso a miembros

El encapsulamiento es uno de los cuatro pilares fundamentales de la Programación Orientada a Objetos (POO). Se refiere a la restricción del acceso a ciertos componentes de un objeto, asegurando que solo se pueda acceder a ciertos atributos y métodos de un objeto desde fuera de la clase de la manera deseada.

### Pregunta problema: 
Imagina que estás diseñando una clase Autor para una biblioteca. ¿Cómo puedes garantizar que ciertos atributos, como el nombre, la nacionalidad o la fecha de nacimiento, no sean modificados accidentalmente o accedidos de manera inapropiada?

## Conceptos clave

_Encapsulamiento:_ Es la técnica de hacer que los campos de una clase sean privados o protegidos y proporcionar acceso a través de métodos públicos.

Diferencias entre "_" y "__" en Python:

_: Se utiliza para indicar que un atributo o método es protegido. Aunque técnicamente aún es accesible desde fuera de la clase, por convención se entiende que no debe ser accedido directamente. Sin embargo, esta protección es solo una convención y no impide realmente el acceso.

__: Se utiliza para indicar que un atributo o método es privado. Esto modifica el nombre del atributo/método de manera que es más difícil de acceder desde fuera de la clase, proporcionando una capa adicional de seguridad.

## Ejemplos prácticos

### Clase Autor con atributos públicos:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def mostrar_autor(self):
        print(f"Nombre: {self.nombre}")
        print(f"Nacionalidad: {self.nacionalidad}")
```

```python

class Autor:
    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def mostrar_autor(self):
        print(f"Nombre: {self.nombre}")
        print(f"Nacionalidad: {self.nacionalidad}")


autor1 = Autor("Mario Benedetti", "Uruguayo")
autor2 = Autor("Isabel Allende", "Chilena")

print("===== AUTORES =====")
autor1.mostrar_autor()

print()

autor2.mostrar_autor()

print("\n===== MODIFICAR ATRIBUTOS =====")

autor1.nacionalidad = "Uruguay"
autor2.nombre = "Isabel Angélica Allende"

autor1.mostrar_autor()

print()

autor2.mostrar_autor()
```

Este diseño, aunque funcional, no protege los atributos nombre y nacionalidad de modificaciones no deseadas.

### Clase Autor con atributos protegidos:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self._nombre = nombre
        self._nacionalidad = nacionalidad
```

```python

class Autor:
    def __init__(self, nombre, nacionalidad):
        self._nombre = nombre
        self._nacionalidad = nacionalidad

    def mostrar_autor(self):
        print("Nombre:", self._nombre)
        print("Nacionalidad:", self._nacionalidad)


autor1 = Autor("Mario Benedetti", "Uruguayo")
autor2 = Autor("Isabel Allende", "Chilena")

print("===== AUTORES =====")
autor1.mostrar_autor()

print()

autor2.mostrar_autor()
```

Aquí, hemos marcado los atributos como protegidos usando un solo guion bajo. Esto indica que no deben ser accedidos directamente, pero aún es posible hacerlo.

### Clase Autor con atributos privados y métodos getter y setter:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad

    def get_nombre(self):
        return self.__nombre

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def get_nacionalidad(self):
        return self.__nacionalidad

    def set_nacionalidad(self, nacionalidad):
        self.__nacionalidad = nacionalidad
```

Con los atributos marcados como privados, ahora es mucho más difícil acceder a ellos desde fuera de la clase. Los métodos getter y setter nos permiten controlar cómo se accede y modifica la información.

```python

class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad

    def get_nombre(self):
        return self.__nombre

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def get_nacionalidad(self):
        return self.__nacionalidad

    def set_nacionalidad(self, nacionalidad):
        self.__nacionalidad = nacionalidad


autor1 = Autor("Mario Benedetti", "Uruguayo")

print("===== DATOS ORIGINALES =====")
print("Nombre:", autor1.get_nombre())
print("Nacionalidad:", autor1.get_nacionalidad())

autor1.set_nombre("Mario Orlando Benedetti")
autor1.set_nacionalidad("Uruguay")

print("\n===== DATOS MODIFICADOS =====")
print("Nombre:", autor1.get_nombre())
print("Nacionalidad:", autor1.get_nacionalidad())
```

```python
#DESAFÍO 52
class Libro:
    def __init__(self, titulo, autor, isbn):
        self.__titulo = titulo
        self.__autor = autor
        self.__isbn = isbn

    def get_titulo(self):
        return self.__titulo

    def set_titulo(self, titulo):
        self.__titulo = titulo

    def get_autor(self):
        return self.__autor

    def set_autor(self, autor):
        self.__autor = autor

    def get_isbn(self):
        return self.__isbn

    def set_isbn(self, isbn):
        self.__isbn = isbn


libro1 = Libro("La tregua", "Mario Benedetti", "9788432215391")

print("Título:", libro1.get_titulo())
print("Autor:", libro1.get_autor())
print("ISBN:", libro1.get_isbn())

libro1.set_titulo("Gracias por el fuego")

print("\nTítulo modificado:", libro1.get_titulo())
```

```python
#DESAFÍO 53
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad
        self.__libros = []

    def agregar_libro(self, libro):
        self.__libros.append(libro)

    def mostrar_libros(self):
        print("Libros de", self.__nombre)
        for libro in self.__libros:
            print("-", libro)


autor1 = Autor("Mario Benedetti", "Uruguayo")

autor1.agregar_libro("La tregua")
autor1.agregar_libro("Gracias por el fuego")
autor1.agregar_libro("Primavera con una esquina rota")

autor1.mostrar_libros()
```

```python
#DESAFÍO 54

class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad

    @property
    def nombre(self):
        return self.__nombre

    @nombre.setter
    def nombre(self, nuevo_nombre):
        if nuevo_nombre != "":
            self.__nombre = nuevo_nombre
        else:
            print("El nombre no puede estar vacío.")

    @property
    def nacionalidad(self):
        return self.__nacionalidad

    @nacionalidad.setter
    def nacionalidad(self, nueva_nacionalidad):
        if nueva_nacionalidad != "":
            self.__nacionalidad = nueva_nacionalidad
        else:
            print("La nacionalidad no puede estar vacía.")


autor1 = Autor("Gabriel García Márquez", "Colombiana")

print("===== DATOS ORIGINALES =====")
print("Nombre:", autor1.nombre)
print("Nacionalidad:", autor1.nacionalidad)

autor1.nombre = "Julio Cortázar"
autor1.nacionalidad = "Argentina"

print("\n===== DATOS MODIFICADOS =====")
print("Nombre:", autor1.nombre)
print("Nacionalidad:", autor1.nacionalidad)

print("\n===== PRUEBA CON VALORES VACÍOS =====")
autor1.nombre = ""
autor1.nacionalidad = ""

print("\n===== DATOS FINALES =====")
print("Nombre:", autor1.nombre)
print("Nacionalidad:", autor1.nacionalidad)
```

```python
#DESAFÍO 55
class Autor:
    def __init__(self, nombre):
        self.__nombre = nombre
        self.__libros = []

    def agregar_libro(self, libro):
        self.__libros.append(libro)

    def obtener_libros(self):
        return self.__libros.copy()


def buscar_titulos(autor):
    return autor.obtener_libros()


autor1 = Autor("Mario Benedetti")

autor1.agregar_libro("La tregua")
autor1.agregar_libro("Gracias por el fuego")
autor1.agregar_libro("Primavera con una esquina rota")

titulos = buscar_titulos(autor1)

print("===== LIBROS DEL AUTOR =====")

for titulo in titulos:
    print("-", titulo)
```

```python
#DESAFÍO 56
class Autor:
    def __init__(self, nombre):
        self.__nombre = nombre
        self.__libros = []

    def agregar_libro(self, libro):
        self.__libros.append(libro)

    def get_nombre(self):
        return self.__nombre

    def cantidad_libros(self):
        return len(self.__libros)


def autor_con_mas_libros(autores):
    if not autores:
        return None

    mayor = autores[0]

    for autor in autores:
        if autor.cantidad_libros() > mayor.cantidad_libros():
            mayor = autor

    return mayor


autor1 = Autor("Mario Benedetti")
autor1.agregar_libro("La tregua")
autor1.agregar_libro("Gracias por el fuego")
autor1.agregar_libro("Primavera con una esquina rota")

autor2 = Autor("Gabriel García Márquez")
autor2.agregar_libro("Cien años de soledad")
autor2.agregar_libro("El amor en los tiempos del cólera")

autor3 = Autor("Isabel Allende")
autor3.agregar_libro("La casa de los espíritus")

autores = [autor1, autor2, autor3]

resultado = autor_con_mas_libros(autores)

if resultado is not None:
    print("Autor con más libros:", resultado.get_nombre())
    print("Cantidad de libros:", resultado.cantidad_libros())
else:
    print("No hay autores registrados.")
```
