# Polimorfismo

El polimorfismo es otro concepto fundamental en la Programación Orientada a Objetos que permite que un objeto pueda adoptar múltiples formas. En el contexto de la herencia, el polimorfismo permite que un objeto de la subclase pueda ser tratado como un objeto de la clase base, lo que facilita la escritura de código más genérico y reutilizable.

## Ejemplo de polimorfismo en la biblioteca

Vamos a ver cómo podemos aplicar el polimorfismo en nuestro ejemplo de la biblioteca. Supongamos que queremos tener una función que imprima la información de cualquier autor, sin importar si es un Autor, un Escritor, un EscritorAcademico, etc.

```python
def imprimir_informacion_autor(autor):
    print("Nombre:", autor.nombre)
    if isinstance(autor, Escritor):
        print("Género Literario:", autor.genero)
    if isinstance(autor, Academico):
        print("Universidad:", autor.universidad)
```

```python

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre


class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero


class EscritorAcademico(Escritor):
    def __init__(self, nombre, genero, universidad):
        super().__init__(nombre, genero)
        self.universidad = universidad


def imprimir_informacion_autor(autor):
    print("Nombre:", autor.nombre)

    if isinstance(autor, Escritor):
        print("Género literario:", autor.genero)

    if isinstance(autor, EscritorAcademico):
        print("Universidad:", autor.universidad)


autor1 = Autor("Julio Cortázar")
escritor1 = Escritor("Mario Benedetti", "Poesía")
academico1 = EscritorAcademico("Umberto Eco", "Ensayo", "Universidad de Bolonia")

print("--- AUTOR ---")
imprimir_informacion_autor(autor1)

print("\n--- ESCRITOR ---")
imprimir_informacion_autor(escritor1)

print("\n--- ESCRITOR ACADÉMICO ---")
imprimir_informacion_autor(academico1)
```

Ahora, podemos pasar cualquier objeto de una clase que herede de Autor a esta función, y se imprimirá la información correspondiente:

```python
autor = Autor("Julio Cortázar")
escritor_academico = EscritorAcademico("Umberto Eco", "Novela Histórica", "Universidad de Bolonia")

imprimir_informacion_autor(autor)
imprimir_informacion_autor(escritor_academico)
```

```python

autores = [
    Autor("Isabel Allende"),
    Escritor("Pablo Neruda", "Poesía"),
    EscritorAcademico("Umberto Eco", "Novela", "Universidad de Bolonia"),
    Escritor("Horacio Quiroga", "Cuento")
]

print("===== LISTA DE AUTORES =====")

for autor in autores:
    imprimir_informacion_autor(autor)
    print("--------------------")
```

## Sobrescritura de métodos

El polimorfismo también nos permite sobrescribir métodos en las subclases. Por ejemplo, podríamos tener un método informacion() en la clase Autor y sobrescribirlo en las subclases para que devuelva información adicional:

```python
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
    
    def informacion(self):
        return f"Nombre: {self.nombre}"

class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero
    
    def informacion(self):
        return f"{super().informacion()} - Género Literario: {self.genero}"

# Instanciamos un objeto de la clase Escritor para Mario Benedetti
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.informacion())
```

```python

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre

    def informacion(self):
        return f"Nombre: {self.nombre}"


class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero

    def informacion(self):
        return f"{super().informacion()} - Género literario: {self.genero}"


autor1 = Autor("Julio Cortázar")
escritor1 = Escritor("Mario Benedetti", "Poesía")
escritor2 = Escritor("Gabriel García Márquez", "Realismo mágico")

print("--- AUTOR ---")
print(autor1.informacion())

print("\n--- ESCRITORES ---")
print(escritor1.informacion())
print(escritor2.informacion())
```

El polimorfismo, junto con la herencia, nos permite escribir código más flexible y reutilizable, al permitirnos tratar objetos de subclases como si fueran objetos de la clase base y sobrescribir métodos para añadir funcionalidades específicas a las subclases.

## Ejemplos adicionales de polimorfismo
### Polimorfismo con métodos de clase

```python
class Animal:
    def sonido(self):
        return "Algunos animales hacen sonidos"

class Perro(Animal):
    def sonido(self):
        return "Guau Guau"

class Gato(Animal):
    def sonido(self):
        return "Miau Miau"
    
class Loro(Animal):
    def sonido(self):
        return "Prr Prr"

animales = [Perro(), Gato(), Loro(), Animal()]

for animal in animales:
    print(animal.sonido())
```

```python

class Animal:
    def sonido(self):
        return "Los animales hacen diferentes sonidos"


class Perro(Animal):
    def sonido(self):
        return "Guau guau"


class Gato(Animal):
    def sonido(self):
        return "Miau miau"


class Vaca(Animal):
    def sonido(self):
        return "Muuu"


class Pato(Animal):
    def sonido(self):
        return "Cuac cuac"


animales = [Perro(), Gato(), Vaca(), Pato(), Animal()]

print("SONIDOS DE LOS ANIMALES")

for animal in animales:
    print(animal.sonido())
```

```python
#DESAFÍO 62
class Musico:
    def instrumento(self):
        return "Toca un instrumento musical"


class Guitarrista(Musico):
    def instrumento(self):
        return "Toca la guitarra"


class Baterista(Musico):
    def instrumento(self):
        return "Toca la batería"


musicos = [Guitarrista(), Baterista(), Musico()]

for musico in musicos:
    print(musico.instrumento())
```

```python
#DESAFÍO 63

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre

    def biografia(self):
        return f"Autor: {self.nombre}"


class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero

    def biografia(self):
        return f"{super().biografia()} - Género: {self.genero}"


escritor1 = Escritor("Mario Benedetti", "Poesía")

print("Biografía del escritor:")
print(escritor1.biografia())

print("\nBiografía de la clase Autor:")
print(super(Escritor, escritor1).biografia())
```

```python
#DESAFÍO 64

class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def mostrar_info(self):
        print("Título:", self.titulo)
        print("Autor:", self.autor)


class LibroEspecializado(Libro):
    def __init__(self, titulo, autor, campo_estudio, nivel):
        super().__init__(titulo, autor)
        self.campo_estudio = campo_estudio
        self.nivel = nivel

    def mostrar_info(self):
        super().mostrar_info()
        print("Campo de estudio:", self.campo_estudio)
        print("Nivel:", self.nivel)


libro1 = LibroEspecializado("Introducción a Python", "Ana López", "Programación", "Básico")

libro1.mostrar_info()
```

```python
#DESAFÍO 65

import math


class Figura:
    def area(self):
        pass


class Circulo(Figura):
    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return math.pi * self.radio ** 2


class Cuadrado(Figura):
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2


circulo = Circulo(3)
cuadrado = Cuadrado(6)

figuras = [circulo, cuadrado]

for figura in figuras:
    print("Figura:", figura.__class__.__name__)
    print("Área:", round(figura.area(), 2))
    print()
```

```python
#DESAFÍO 66

class Operacion:
    def resultado(self):
        return 0


class Suma(Operacion):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def resultado(self):
        return self.a + self.b


class Multiplicacion(Operacion):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def resultado(self):
        return self.a * self.b


operaciones = [Suma(8, 4), Multiplicacion(8, 4)]

for operacion in operaciones:
    print("Operación:", operacion.__class__.__name__)
    print("Resultado:", operacion.resultado())
    print()
```
