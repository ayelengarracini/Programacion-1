# Introducción a los Tipos Abstractos de Datos (TAD)

## Concepto de TAD

Un Tipo Abstracto de Datos (TAD) es una especificación de un conjunto de datos y las operaciones que se pueden realizar con esos datos. El TAD se centra en qué operaciones existen y no en cómo se implementan. Es una herramienta conceptual que nos permite abstraernos de los detalles y pensar en términos de operaciones y datos.

### Ejemplo de TAD: Lista


```python
lista = []

def agregar(elemento):
    lista.append(elemento)

def eliminar(elemento):
    if elemento in lista:
        lista.remove(elemento)

def buscar(elemento):
    return elemento in lista
```

```python

# TAD: LISTA

# Función para agregar elementos
def agregar(lista, elemento):
    nueva_lista = lista.copy()
    nueva_lista.append(elemento)
    return nueva_lista

# Función para eliminar elementos
def eliminar(lista, elemento):
    nueva_lista = lista.copy()
    if elemento in nueva_lista:
        nueva_lista.remove(elemento)
    return nueva_lista

# Función para buscar elementos
def buscar(lista, elemento):
    return elemento in lista


# Creamos una lista de frutas
frutas = []

# Agregamos elementos
frutas = agregar(frutas, "manzana")
frutas = agregar(frutas, "banana")
frutas = agregar(frutas, "pera")

print("Lista original:", frutas)

# Buscamos un elemento
print("¿Hay manzanas?", buscar(frutas, "manzana"))

# Eliminamos un elemento
frutas = eliminar(frutas, "banana")
print("Lista modificada:", frutas)

# Experimento: agregar otra fruta
frutas = agregar(frutas, "naranja")
print("Lista final:", frutas)
```

## Desventajas de este enfoque

1. **Falta de encapsulamiento**: En el ejemplo proporcionado, tanto las operaciones como los datos están expuestos y pueden ser modificados directamente desde fuera del TAD. Esto viola el principio de encapsulamiento, que es fundamental para el diseño de software robusto y mantenible.

2. **Dificultad en reutilización**: Con el diseño actual, si quisieras tener múltiples listas y aplicar operaciones sobre ellas, tendrías que crear funciones separadas o modificar las existentes, lo que no es práctico.

3. **Problemas de extensibilidad**: Cambiar la implementación interna (por ejemplo, cambiar de una lista a un árbol binario para mejorar la eficiencia) requeriría modificar todas las funciones que interactúan con la estructura de datos. Esto hace que el código sea difícil de mantener y de extender con nuevas funcionalidades.

4. **Falta de abstracción**: Aunque el TAD debería ayudarnos a abstraernos de los detalles de implementación, el enfoque actual no lo hace de manera efectiva. Esto podría llevar a errores si no se comprenden completamente las implementaciones subyacentes.

### Desafío 46: Estructuras de datos anidadas
Considera un sistema de manejo de inventario para una cadena de tiendas minoristas. Tienes que tratar con datos de productos, tiendas, empleados, y transacciones, donde cada tienda podría tener múltiples productos y empleados. Gestionar estas relaciones con TADs podría ser ineficiente y propenso a errores.


```python

# DESAFÍO 44

# Función para agregar un libro
def agregar_libro(biblioteca, titulo):
    nueva_biblioteca = biblioteca.copy()
    nueva_biblioteca.append(titulo)
    print(f'Libro "{titulo}" agregado correctamente.')
    return nueva_biblioteca


# Función para eliminar un libro
def eliminar_libro(biblioteca, titulo):
    nueva_biblioteca = biblioteca.copy()

    if titulo in nueva_biblioteca:
        nueva_biblioteca.remove(titulo)
        print(f'Libro "{titulo}" eliminado correctamente.')
    else:
        print(f'El libro "{titulo}" no se encuentra en la biblioteca.')

    return nueva_biblioteca


# Función para buscar un libro
def buscar_libro(biblioteca, titulo):
    if titulo in biblioteca:
        print(f'El libro "{titulo}" está disponible.')
    else:
        print(f'El libro "{titulo}" no está disponible.')


# Función para mostrar todos los libros
def mostrar_libros(biblioteca):
    if biblioteca:
        print("\nLibros disponibles:")
        for libro in biblioteca:
            print(f"- {libro}")
    else:
        print("No hay libros disponibles.")


# PRUEBA DEL TAD

biblioteca = []

# Agregar libros
biblioteca = agregar_libro(biblioteca, "Cien años de soledad")
biblioteca = agregar_libro(biblioteca, "El Principito")
biblioteca = agregar_libro(biblioteca, "Don Quijote de la Mancha")
biblioteca = agregar_libro(biblioteca, "1984")

# Buscar un libro
print("\n--- BÚSQUEDA ---")
buscar_libro(biblioteca, "El Principito")

# Eliminar un libro
print("\n--- ELIMINACIÓN ---")
biblioteca = eliminar_libro(biblioteca, "1984")

# Mostrar la biblioteca
print("\n--- BIBLIOTECA FINAL ---")
mostrar_libros(biblioteca)
```

```python
# DESAFÍO 45

def agregar(biblioteca, libro):
    nueva_biblioteca = biblioteca.copy()
    nueva_biblioteca.append(libro)
    return nueva_biblioteca


def eliminar(biblioteca, libro):
    nueva_biblioteca = biblioteca.copy()

    if libro in nueva_biblioteca:
        nueva_biblioteca.remove(libro)
        print(f'"{libro}" eliminado correctamente.')
    else:
        print(f'"{libro}" no se encuentra en la biblioteca.')

    return nueva_biblioteca


def buscar(biblioteca, libro):
    if libro in biblioteca:
        print(f'"{libro}" está disponible.')
    else:
        print(f'"{libro}" no está disponible.')


def mostrar(biblioteca):
    if biblioteca:
        for libro in biblioteca:
            print(f"- {libro}")
    else:
        print("La biblioteca está vacía.")


biblioteca1 = []
biblioteca2 = []

biblioteca1 = agregar(biblioteca1, "Cien años de soledad")
biblioteca1 = agregar(biblioteca1, "El Principito")

biblioteca2 = agregar(biblioteca2, "1984")
biblioteca2 = agregar(biblioteca2, "Don Quijote de la Mancha")

buscar(biblioteca1, "El Principito")

biblioteca2 = eliminar(biblioteca2, "1984")

print("\n--- Biblioteca 1 ---")
mostrar(biblioteca1)

print("\n--- Biblioteca 2 ---")
mostrar(biblioteca2)
```

```python

# DESAFÍO 46

tiendas = {
    "Tienda Centro": {
        "productos": {
            "Manzanas": 20,
            "Bananas": 15,
            "Peras": 10
        },
        "empleados": ["Ana", "Pedro"],
        "transacciones": []
    },

    "Tienda Norte": {
        "productos": {
            "Manzanas": 12,
            "Naranjas": 25
        },
        "empleados": ["Lucía", "Juan"],
        "transacciones": []
    }
}


def mostrar_tienda(nombre, datos):
    print("\n---", nombre, "---")

    print("Productos:")
    for producto, cantidad in datos["productos"].items():
        print(producto, ":", cantidad)

    print("Empleados:")
    for empleado in datos["empleados"]:
        print("-", empleado)

    print("Transacciones:", datos["transacciones"])


def vender_producto(datos, producto, cantidad):
    nueva_tienda = {
        "productos": datos["productos"].copy(),
        "empleados": datos["empleados"].copy(),
        "transacciones": datos["transacciones"].copy()
    }

    if producto in nueva_tienda["productos"]:
        if nueva_tienda["productos"][producto] >= cantidad:
            nueva_tienda["productos"][producto] -= cantidad
            nueva_tienda["transacciones"].append(
                f"Venta de {cantidad} unidades de {producto}"
            )
            print("Venta realizada correctamente.")
        else:
            print("No hay suficiente stock.")
    else:
        print("El producto no existe.")

    return nueva_tienda


print("===== INVENTARIO INICIAL =====")

for nombre, datos in tiendas.items():
    mostrar_tienda(nombre, datos)


print("\n===== REALIZAR UNA VENTA =====")

tiendas_actualizadas = tiendas.copy()

tiendas_actualizadas["Tienda Centro"] = vender_producto(
    tiendas["Tienda Centro"], "Manzanas", 5
)


print("\n===== INVENTARIO ACTUALIZADO =====")

mostrar_tienda(
    "Tienda Centro",
    tiendas_actualizadas["Tienda Centro"]
)


print("\n===== OTRO EXPERIMENTO =====")

tiendas_actualizadas["Tienda Norte"] = vender_producto(
    tiendas["Tienda Norte"], "Naranjas", 3
)

mostrar_tienda(
    "Tienda Norte",
    tiendas_actualizadas["Tienda Norte"]
)

print("\n===== FIN DEL PROGRAMA =====")
```
