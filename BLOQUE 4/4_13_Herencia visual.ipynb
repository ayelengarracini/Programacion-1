# Introducción a la Herencia

En nuestro viaje por el mundo de la Programación Orientada a Objetos (POO), hemos creado clases que representan entidades en una biblioteca, como `Autor` y `Libro`. Pero, ¿qué pasa si queremos crear nuevas clases que compartan características con las ya existentes, pero con algunas diferencias o adicionales? ¡Aquí es donde entra en juego la herencia!

### Pregunta problema: 
¿Cómo podemos extender nuestras clases existentes para representar diferentes tipos de autores o libros, sin tener que reescribir todo el código?



```python
# Definimos la clase base Autor
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
```

## Conceptos clave de la herencia

La **herencia** permite que una clase (llamada subclase o clase hija) herede atributos y métodos de otra clase (llamada superclase o clase base). La herencia facilita la reutilización y extensión del código.

- **Herencia simple:** Una clase deriva de una única clase base.
- **Herencia múltiple:** Una clase puede derivar de varias clases base.
- **super():** Función que permite llamar a métodos de la clase base desde la clase derivada.

```python
# Creamos una subclase Escritor que hereda de Autor
class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero
```

```python

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre

    def mostrar_autor(self):
        print("Nombre:", self.nombre)


class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero

    def mostrar_escritor(self):
        self.mostrar_autor()
        print("Género literario:", self.genero)


escritor1 = Escritor("Mario Benedetti", "Poesía")
escritor2 = Escritor("Gabriel García Márquez", "Realismo mágico")

print("===== ESCRITOR 1 =====")
escritor1.mostrar_escritor()

print("\n===== ESCRITOR 2 =====")
escritor2.mostrar_escritor()
```

## Ejemplos prácticos de Herencia

Vamos a extender nuestra clase `Autor` para crear una nueva clase `Escritor` que tenga un atributo adicional: el género literario principal en el que escribe.


```python
# Instanciamos un objeto de la clase Escritor
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.nombre, escritor.genero)
```

```python

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre


class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero


escritor1 = Escritor("Mario Benedetti", "Poesía")
escritor2 = Escritor("Isabel Allende", "Novela")

print("===== ESCRITORES =====")
print(escritor1.nombre, "-", escritor1.genero)
print(escritor2.nombre, "-", escritor2.genero)
```

## Herencia Múltiple

En Python, una clase puede heredar de varias clases base, lo que se conoce como herencia múltiple. Aunque puede ser útil, también puede llevar a complicaciones si varias clases base tienen atributos o métodos con el mismo nombre.


```python
# Definimos una segunda clase base
class Academico:
    def __init__(self, universidad):
        self.universidad = universidad

# Creamos una clase que hereda de Escritor y Academico
class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        Escritor.__init__(self, nombre, genero)
        Academico.__init__(self, universidad)
```

```python

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre


class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero


class Academico:
    def __init__(self, universidad):
        self.universidad = universidad


class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        Escritor.__init__(self, nombre, genero)
        Academico.__init__(self, universidad)

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Género:", self.genero)
        print("Universidad:", self.universidad)


escritor1 = EscritorAcademico("Ana Pérez", "Ensayo", "Universidad de la República")
escritor2 = EscritorAcademico("Juan López", "Historia", "Universidad de Buenos Aires")

print("===== ESCRITOR ACADÉMICO 1 =====")
escritor1.mostrar_datos()

print("\n===== ESCRITOR ACADÉMICO 2 =====")
escritor2.mostrar_datos()
```

### Desafío 61:
Crea una jerarquía de clases para representar diferentes tipos de empleados en una biblioteca, utilizando herencia múltiple y composición. Por ejemplo, implementa clases como Empleado, Gerente, Tecnico, y Voluntario, donde Gerente y Tecnico hereden de Empleado, y algunos puedan tener roles adicionales mediante composición con otras clases como Administrador o Mantenimiento.



```python
#DESAFÍO 57
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Nacionalidad:", self.nacionalidad)


class Poeta(Autor):
    def __init__(self, nombre, nacionalidad, tipo_poesia):
        super().__init__(nombre, nacionalidad)
        self.tipo_poesia = tipo_poesia

    def mostrar_datos_poeta(self):
        self.mostrar_datos()
        print("Tipo de poesía:", self.tipo_poesia)


poeta1 = Poeta("Pablo Neruda", "Chilena", "Lírica")
poeta2 = Poeta("Emily Dickinson", "Estadounidense", "Romántica")

print("===== POETA 1 =====")
poeta1.mostrar_datos_poeta()

print("\n===== POETA 2 =====")
poeta2.mostrar_datos_poeta()

print("\n===== MÉTODO HEREDADO =====")
poeta1.mostrar_datos()
```

```python
#DESAFÍO 58

class Usuario:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def mostrar_usuario(self):
        print("Nombre:", self.nombre)
        print("Edad:", self.edad)


class Bibliotecario(Usuario):
    def __init__(self, nombre, edad, seccion, años_experiencia):
        super().__init__(nombre, edad)
        self.seccion = seccion
        self.años_experiencia = años_experiencia

    def mostrar_datos(self):
        self.mostrar_usuario()
        print("Sección:", self.seccion)
        print("Años de experiencia:", self.años_experiencia)


bibliotecario1 = Bibliotecario("Lucía Pérez", 35, "Literatura", 8)
bibliotecario2 = Bibliotecario("Juan Rodríguez", 42, "Historia", 12)

print("===== BIBLIOTECARIO 1 =====")
bibliotecario1.mostrar_datos()

print("\n===== BIBLIOTECARIO 2 =====")
bibliotecario2.mostrar_datos()
```

```python
#DESAFÍO 59

class Libro:
    def __init__(self, titulo, autor, isbn):
        self.titulo = titulo
        self.autor = autor
        self.isbn = isbn


class LibroDigital(Libro):
    def __init__(self, titulo, autor, isbn, formato, tamaño_archivo):
        super().__init__(titulo, autor, isbn)
        self.formato = formato
        self.tamaño_archivo = tamaño_archivo

    def mostrar_informacion(self):
        print("Título:", self.titulo)
        print("Autor:", self.autor)
        print("ISBN:", self.isbn)
        print("Formato:", self.formato)
        print("Tamaño del archivo:", self.tamaño_archivo)


class EBook(LibroDigital):
    def __init__(self, titulo, autor, isbn, formato, tamaño_archivo, enlace_descarga):
        super().__init__(titulo, autor, isbn, formato, tamaño_archivo)
        self.enlace_descarga = enlace_descarga

    def mostrar_informacion(self):
        super().mostrar_informacion()
        print("Enlace de descarga:", self.enlace_descarga)


libro_digital = LibroDigital(
    "El Principito",
    "Antoine de Saint-Exupéry",
    "978-0156012195",
    "PDF",
    "2 MB"
)

ebook = EBook(
    "Harry Potter y la piedra filosofal",
    "J.K. Rowling",
    "978-8478884456",
    "EPUB",
    "5 MB",
    "https://ejemplo.com/descarga"
)

print("===== LIBRO DIGITAL =====")
libro_digital.mostrar_informacion()

print("\n===== EBOOK =====")
ebook.mostrar_informacion()
```

```python
#DESAFÍO 60
class Escritor:
    def __init__(self, nombre, genero, universidad):
        super().__init__(universidad)
        self.nombre = nombre
        self.genero = genero


class Academico:
    def __init__(self, universidad):
        super().__init__()
        self.universidad = universidad


class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        super().__init__(nombre, genero, universidad)

    def publicar_articulo(self, titulo):
        print(self.nombre, "publicó el artículo:", titulo)

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Género:", self.genero)
        print("Universidad:", self.universidad)


escritor1 = EscritorAcademico(
    "Ana Pérez",
    "Ensayo",
    "Universidad de la República"
)

print("===== ESCRITOR ACADÉMICO =====")
escritor1.mostrar_datos()

print("\n===== PUBLICACIÓN =====")
escritor1.publicar_articulo("La importancia de la lectura")
```

```python
#DESAFÍO 61

class Empleado:
    def __init__(self, nombre):
        self.nombre = nombre

    def mostrar_nombre(self):
        print("Empleado:", self.nombre)


class Administrador:
    def administrar(self):
        print("Organizando las tareas de la biblioteca")


class Mantenimiento:
    def reparar(self):
        print("Realizando mantenimiento de equipos")


class Gerente(Empleado):
    def __init__(self, nombre):
        super().__init__(nombre)
        self.administrador = Administrador()

    def trabajar(self):
        self.mostrar_nombre()
        self.administrador.administrar()


class Tecnico(Empleado):
    def __init__(self, nombre):
        super().__init__(nombre)
        self.mantenimiento = Mantenimiento()

    def trabajar(self):
        self.mostrar_nombre()
        self.mantenimiento.reparar()


class Voluntario(Empleado):
    def trabajar(self):
        self.mostrar_nombre()
        print("Ayudando a ordenar los libros")


class Coordinador(Gerente, Tecnico):
    def trabajar(self):
        self.mostrar_nombre()
        self.administrador.administrar()
        self.mantenimiento.reparar()


gerente = Gerente("María")
tecnico = Tecnico("Pedro")
voluntario = Voluntario("Lucía")
coordinador = Coordinador("Sofía")

print("===== GERENTE =====")
gerente.trabajar()

print("\n===== TÉCNICO =====")
tecnico.trabajar()

print("\n===== VOLUNTARIO =====")
voluntario.trabajar()

print("\n===== COORDINADOR =====")
coordinador.trabajar()
```
