## Problema 5: Lectura y Escritura en Archivos
Supongamos que ahora las calificaciones de los alumnos se almacenan en un archivo JSON en lugar de un CSV. El archivo JSON contiene una lista de diccionarios, donde cada diccionario representa a un estudiante y sus calificaciones. Nuestra tarea será leer este archivo, calcular el promedio de las calificaciones y luego escribir ese promedio en un nuevo archivo JSON.

```python
import json

# Leemos el archivo JSON
with open('calificaciones.json', 'r') as archivo_calificaciones:
    estudiantes = json.load(archivo_calificaciones)
    
    suma_total = 0
    contador = 0
    for estudiante in estudiantes:
        suma_total += estudiante['calificacion']
        contador += 1

# Calculamos el promedio
promedio = suma_total / contador if contador > 0 else 0

# Escribimos el promedio de calificaciones en un nuevo archivo JSON
with open('promedio_calificaciones.json', 'w') as archivo_promedio:
    json.dump({"promedio_calificaciones": promedio}, archivo_promedio)

print("El promedio es: ", promedio)
```

```python

# ==========================================
# PROBLEMA 5: LECTURA Y ESCRITURA EN JSON
# ==========================================

import json

# 1. Creamos datos de ejemplo
# Cada estudiante tiene un nombre y una calificacion

estudiantes = [
    {"nombre": "Ana", "calificacion": 9},
    {"nombre": "Juan", "calificacion": 7},
    {"nombre": "Maria", "calificacion": 10},
    {"nombre": "Pedro", "calificacion": 6},
    {"nombre": "Lucia", "calificacion": 8},
    {"nombre": "Sofia", "calificacion": 11}
]

# 2. Guardamos los datos en un archivo JSON
with open("calificaciones.json", "w",
          encoding="utf-8") as archivo:

    json.dump(estudiantes, archivo,
              indent=4, ensure_ascii=False)

print("Archivo calificaciones.json creado correctamente.")

# 3. Leemos el archivo JSON
with open("calificaciones.json", "r",
          encoding="utf-8") as archivo:

    datos = json.load(archivo)

# 4. Calculamos el promedio general
suma_total = 0
contador = 0

for estudiante in datos:
    suma_total += estudiante["calificacion"]
    contador += 1

promedio = suma_total / contador if contador > 0 else 0

# 5. Buscamos las calificaciones maxima y minima
if datos:
    mejor_estudiante = max(
        datos, key=lambda e: e["calificacion"]
    )
    menor_estudiante = min(
        datos, key=lambda e: e["calificacion"]
    )
else:
    mejor_estudiante = None
    menor_estudiante = None

# 6. Mostramos los resultados
print("\n===== REPORTE DE CALIFICACIONES =====")

for estudiante in datos:
    print(f"{estudiante['nombre']:<12} "
          f"Nota: {estudiante['calificacion']}")

print("\n===== ESTADISTICAS =====")

print("Cantidad de estudiantes:", contador)
print(f"Promedio general: {promedio:.2f}")

if mejor_estudiante is not None:
    print("Calificacion mas alta:",
          mejor_estudiante["nombre"],
          "-", mejor_estudiante["calificacion"])

    print("Calificacion mas baja:",
          menor_estudiante["nombre"],
          "-", menor_estudiante["calificacion"])

# 7. Preparamos el reporte para guardar
reporte = {
    "cantidad_estudiantes": contador,
    "promedio_calificaciones": round(promedio, 2),
    "calificacion_maxima": (
        mejor_estudiante if mejor_estudiante else None
    ),
    "calificacion_minima": (
        menor_estudiante if menor_estudiante else None
    )
}

# 8. Guardamos el promedio y las estadisticas
with open("promedio_calificaciones.json", "w",
          encoding="utf-8") as archivo:

    json.dump(reporte, archivo,
              indent=4, ensure_ascii=False)

print("\nArchivo promedio_calificaciones.json creado.")
print("===== FIN DEL PROGRAMA =====")
```
