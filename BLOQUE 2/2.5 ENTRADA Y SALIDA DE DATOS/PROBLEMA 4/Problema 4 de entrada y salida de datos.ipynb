## Problema 4: Lectura y Escritura en Archivos
Supongamos que somos profesores y necesitamos procesar las calificaciones de nuestros alumnos. Los datos se almacenan en un archivo CSV, y nuestro trabajo es leer ese archivo, calcular el promedio de las calificaciones y luego escribir ese promedio en un nuevo archivo.

input() y print() no son adecuados aquí porque estamos trabajando con archivos y no con entradas de usuario o impresiones en la consola.

```python
import csv

# Leemos el archivo CSV
with open('calificaciones.csv', 'r') as archivo_calificaciones:
    lector = csv.reader(archivo_calificaciones)
    next(lector)  # Saltamos la cabecera del archivo
    
    suma_total = 0
    contador = 0
    for fila in lector:
        suma_total += float(fila[1])  # Asumimos que la calificación está en la segunda columna
        contador += 1

# Calculamos el promedio
promedio = suma_total / contador if contador else 0
print("El promedio es: ",promedio)

# Escribimos el promedio de calificaciones en un nuevo archivo
with open('promedio_calificaciones.txt', 'w') as archivo_promedio:
    archivo_promedio.write(f"El promedio de las calificaciones es: {promedio}\n")
```

```python

# ==========================================
# PROBLEMA 4: LECTURA Y ESCRITURA DE ARCHIVOS
# ==========================================

import csv

# 1. Creamos una lista de estudiantes
# y sus calificaciones de ejemplo

estudiantes = [
    ["Nombre", "Calificacion"],
    ["Ana", 9],
    ["Juan", 7],
    ["Maria", 10],
    ["Pedro", 6],
    ["Lucia", 8],
    ["Sofia", 11]
]

# 2. Creamos el archivo CSV
with open("calificaciones.csv", "w",
          newline="", encoding="utf-8") as archivo:

    escritor = csv.writer(archivo)
    escritor.writerows(estudiantes)

print("Archivo calificaciones.csv creado correctamente.")


# 3. Leemos el archivo CSV
calificaciones = []
nombres = []

with open("calificaciones.csv", "r",
          encoding="utf-8") as archivo:

    lector = csv.reader(archivo)
    next(lector)  # Saltamos la cabecera

    for fila in lector:
        nombres.append(fila[0])
        calificaciones.append(float(fila[1]))


# 4. Calculamos los resultados
suma_total = sum(calificaciones)
cantidad = len(calificaciones)

promedio = suma_total / cantidad if cantidad > 0 else 0

nota_maxima = max(calificaciones) if cantidad > 0 else 0
nota_minima = min(calificaciones) if cantidad > 0 else 0


# 5. Mostramos los resultados
print("\n===== REPORTE DE CALIFICACIONES =====")

for nombre, nota in zip(nombres, calificaciones):
    print(f"{nombre:<12} Nota: {nota:.1f}")

print("\n===== ESTADISTICAS =====")

print(f"Cantidad de estudiantes: {cantidad}")
print(f"Suma de calificaciones: {suma_total:.2f}")
print(f"Promedio general: {promedio:.2f}")
print(f"Calificacion mas alta: {nota_maxima:.2f}")
print(f"Calificacion mas baja: {nota_minima:.2f}")


# 6. Guardamos los resultados en un archivo TXT
with open("promedio_calificaciones.txt", "w",
          encoding="utf-8") as archivo:

    archivo.write("REPORTE DE CALIFICACIONES\n")
    archivo.write("========================\n")

    for nombre, nota in zip(nombres, calificaciones):
        archivo.write(f"{nombre}: {nota:.1f}\n")

    archivo.write("\n")
    archivo.write(f"Cantidad de estudiantes: {cantidad}\n")
    archivo.write(f"Promedio general: {promedio:.2f}\n")
    archivo.write(f"Nota maxima: {nota_maxima:.2f}\n")
    archivo.write(f"Nota minima: {nota_minima:.2f}\n")


print("\nArchivo promedio_calificaciones.txt creado correctamente.")
print("===== FIN DEL PROGRAMA =====")
```
