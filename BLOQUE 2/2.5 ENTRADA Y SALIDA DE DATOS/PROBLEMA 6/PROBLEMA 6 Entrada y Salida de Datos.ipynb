## Problema 6: Interactuando con una API web
Imagina que estás escribiendo un programa que necesita interactuar con una API web para obtener datos en tiempo real, como el precio actual de una criptomoneda.

input() y print() no son suficientes aquí porque necesitamos enviar una solicitud HTTP a una API web y procesar la respuesta JSON que recibimos.

```python
import requests
import json

# Hacemos una solicitud GET a la API de CoinDesk para obtener el precio actual de Bitcoin
respuesta = requests.get("https://api.coindesk.com/v1/bpi/currentprice/BTC.json")

# Convertimos la respuesta a JSON
datos = respuesta.json()

# Obtenemos el precio actual de Bitcoin en USD
precio_bitcoin = datos['bpi']['USD']['rate']

# Imprimimos el precio actual
print(f"El precio actual de Bitcoin es: {precio_bitcoin} USD")
```

```python

# ==========================================
# PROBLEMA 6: INTERACTUANDO CON UNA API WEB
# ==========================================

import requests
import json

print("===== CONSULTA DE CRIPTOMONEDAS =====")

# 1. Definimos la direccion de la API
url = "https://api.coingecko.com/api/v3/simple/price"

# 2. Indicamos los datos que queremos consultar
parametros = {
    "ids": "bitcoin",
    "vs_currencies": "usd,eur"
}

# 3. Realizamos la solicitud a la API
try:
    respuesta = requests.get(
        url,
        params=parametros,
        timeout=10
    )

    # Verificamos que la solicitud sea correcta
    respuesta.raise_for_status()

    # 4. Convertimos la respuesta a datos de Python
    datos = respuesta.json()

    # 5. Obtenemos los precios de Bitcoin
    precio_usd = datos["bitcoin"]["usd"]
    precio_eur = datos["bitcoin"]["eur"]

    # 6. Mostramos los resultados
    print("\n===== PRECIO DE BITCOIN =====")

    print(f"Bitcoin en dolares: ${precio_usd:,.2f} USD")
    print(f"Bitcoin en euros: {precio_eur:,.2f} EUR")

    # 7. Calculamos el valor de distintas cantidades
    print("\n===== CALCULADORA DE BITCOIN =====")

    cantidad_bitcoin = 0.05

    valor_usd = cantidad_bitcoin * precio_usd
    valor_eur = cantidad_bitcoin * precio_eur

    print(f"Cantidad de Bitcoin: {cantidad_bitcoin}")
    print(f"Valor en dolares: ${valor_usd:,.2f}")
    print(f"Valor en euros: {valor_eur:,.2f}")

    # 8. Guardamos los datos en un archivo JSON
    reporte = {
        "criptomoneda": "Bitcoin",
        "precio_usd": precio_usd,
        "precio_eur": precio_eur,
        "cantidad_bitcoin": cantidad_bitcoin,
        "valor_total_usd": round(valor_usd, 2),
        "valor_total_eur": round(valor_eur, 2)
    }

    with open("precio_bitcoin.json", "w",
              encoding="utf-8") as archivo:

        json.dump(
            reporte,
            archivo,
            indent=4,
            ensure_ascii=False
        )

    print("\nArchivo precio_bitcoin.json creado correctamente.")

# 9. Controlamos posibles errores
except requests.exceptions.RequestException as error:
    print("\nError al conectar con la API.")
    print("Detalle:", error)

except (KeyError, ValueError, TypeError) as error:
    print("\nError al procesar los datos recibidos.")
    print("Detalle:", error)

except OSError as error:
    print("\nError al guardar el archivo.")
    print("Detalle:", error)

print("\n===== FIN DEL PROGRAMA =====")
```
