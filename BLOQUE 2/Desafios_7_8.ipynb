## Desafío 7: Convertidor de temperaturas

Solicitar una temperatura en Celsius y convertirla a Fahrenheit y Kelvin mediante asignaciones directas.


```python
# Desafio 7: Convertidor de temperaturas
# Solicitamos una temperatura en grados Celsius.
celsius = float(input("Ingresa la temperatura en Celsius: "))

# Convertimos directamente al asignar las variables.
fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

# Mostramos los resultados con dos decimales.
print(f"Celsius: {celsius:.2f} °C")
print(f"Fahrenheit: {fahrenheit:.2f} °F")
print(f"Kelvin: {kelvin:.2f} K")
```

**Prueba sugerida:** ingresa `25`. Debes obtener `77.00 °F` y `298.15 K`.


## Desafío 8: Verificar múltiplos de varios números

Solicitar un número entero y comprobar si es múltiplo de 2, 3, 5, 7, 9, 10 y 11 usando `%`.


```python
# Desafio 8: Verificar multiplos de varios numeros
# Solicitamos un numero entero.
numero = int(input("Ingresa un numero entero: "))

# Divisores que debemos comprobar.
divisores = [2, 3, 5, 7, 9, 10, 11]

# El resto cero indica que el numero es multiplo.
for divisor in divisores:
    if numero % divisor == 0:
        print(f"¿{numero} es multiplo de {divisor}? Si")
    else:
        print(f"¿{numero} es multiplo de {divisor}? No")
```

**Prueba sugerida:** ingresa `30`. Debe indicar **Sí** para 2, 3, 5 y 10, y **No** para 7, 9 y 11.

**Nota:** las celdas usan `input()` porque la consigna pide solicitar los valores. Si tu entorno de cuadernos no admite entrada interactiva, sustituye temporalmente la primera asignación por `celsius = 25` o `numero = 30` para hacer las pruebas.
