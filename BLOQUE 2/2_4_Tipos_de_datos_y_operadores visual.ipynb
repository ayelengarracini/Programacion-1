# Identificadores, constantes y variables

En Python, un identificador es un nombre que se utiliza para identificar una variable, función, clase, módulo u otro objeto. Los identificadores son sensibles a mayúsculas y minúsculas y deben seguir ciertas reglas de nomenclatura. Para nombrar los identificadores en Python, se utilizan letras, números y guiones bajos. Sin embargo, el primer carácter no puede ser un número y no se permiten espacios ni caracteres especiales.

Para garantizar la creación correcta de identificadores en Python, es importante seguir las PEP 8, que son las guías de estilo de Python. Estas proporcionan recomendaciones sobre cómo nombrar variables, funciones y otros objetos de manera consistente y legible para otros programadores. Puedes encontrar las [PEP 8 en el sitio web oficial de Python](https://peps.python.org/pep-0008/). Siguiendo estas pautas, puedes mejorar la claridad y la legibilidad de tu código, lo que facilita su mantenimiento y colaboración con otros desarrolladores.

Por ejemplo, los siguientes son identificadores válidos en Python:

```python
numero
total_venta
ListaDeCompras
edad
```

Mientras que los siguientes no son identificadores válidos:

```python
2d_array  # el primer carácter no puede ser un número
mi identificador  # no se permiten espacios
total$  # no se permiten caracteres especiales
```

*Nota 2: Cabe destacar, que por mas que no es inválido, utilizar los caracteres "l" (ele minúscula), "i" (letra i latina mayuscula), "O" (letra o mayúscula) y "0" (número cero), puede causar confusión según el caso en donde se utilicen.*

También se tiene en cuenta que, como la mayoría de lenguajes, Python tiene una lista de palabras reservadas, que no pueden utilizarse como identificadores:

| Palabra reservadas |
|---------------|
| False         |
| True          |
| def           |
| while         |
| if            |
| else          |

Para ver la lista completa utiliza el siguiente codigo en Python:

```python
import keyword

print(keyword.kwlist)
```

Las constantes son valores que no cambian durante la ejecución de un programa. En Python, se utilizan variables para almacenar constantes. Las constantes pueden ser números, cadenas de texto, booleanos o cualquier otro valor inmutable.

Por ejemplo, los siguientes son constantes en Python:

```python
PI = 3.14159
NOMBRE_EMPRESA = "Acme Corporation"
ES_POSITIVO = True
```

Las variables en Python son objetos que se utilizan para almacenar datos. Las variables se crean cuando se les asigna un valor y pueden cambiar durante la ejecución de un programa.

Por ejemplo, el siguiente código crea una variable llamada edad y le asigna el valor 25:

```python
edad = 25
```

En Python, las variables no necesitan ser declaradas antes de usarlas. Python automáticamente asigna un tipo de datos a las variables basándose en el valor que se les asigna. Por ejemplo, si se asigna un número entero a una variable, Python automáticamente la considerará como una variable de tipo int. Si se asigna una cadena de texto a una variable, Python la considerará como una variable de tipo str.

Pero... ¿cómo podemos determinar el tipo de una variable? Si veo "un número" entero, puedo identificar que el tipo de variable es un int o si veo un texto escrito entre (" "), puedo determinar que es una variable str.

```python
edad = 12
nombre = "Anthony"
altura = 1.75

print ("la variable", edad, "es de tipo:", type(edad))
print ("la variable", nombre, "es de tipo:", type(nombre))
print ("la variable", altura, "es de tipo:", type(altura))
```

```python

# ==========================================
# EXPERIMENTANDO CON TIPOS DE VARIABLES
# ==========================================

# 1. Creamos diferentes variables
edad = 20
nombre = "Anthony"
altura = 1.75
estudiante = True
nota = 9.5

print("===== DATOS PERSONALES =====")

print("Nombre:", nombre)
print("Edad:", edad)
print("Altura:", altura)
print("Es estudiante:", estudiante)
print("Nota:", nota)

# 2. Descubrimos el tipo de cada variable
print("\n===== TIPOS DE VARIABLES =====")

print("Nombre:", type(nombre))
print("Edad:", type(edad))
print("Altura:", type(altura))
print("Estudiante:", type(estudiante))
print("Nota:", type(nota))

# 3. Jugamos cambiando los tipos
print("\n===== CAMBIANDO TIPOS =====")

edad_texto = str(edad)
altura_entera = int(altura)
nota_texto = str(nota)

print("Edad como texto:", edad_texto, type(edad_texto))
print("Altura como entero:", altura_entera, type(altura_entera))
print("Nota como texto:", nota_texto, type(nota_texto))

# 4. Hacemos operaciones
print("\n===== OPERACIONES =====")

edad_futura = edad + 5
altura_cm = altura * 100

print("Edad dentro de 5 años:", edad_futura)
print("Altura en centimetros:", altura_cm)

# 5. Comparamos diferentes tipos
print("\n===== COMPARACIONES =====")

numero = 10
texto = "10"

print("Tipo de numero:", type(numero))
print("Tipo de texto:", type(texto))

print("¿Son iguales?", numero == texto)
print("¿Son iguales al convertir?", numero == int(texto))

print("\n===== FIN DEL PROGRAMA =====")
```

Es importante tener en cuenta que las variables en Python son sensibles a mayúsculas y minúsculas. Esto significa que las variables edad y Edad son dos variables diferentes en Python.

Además, las variables en Python pueden ser eliminadas utilizando la palabra clave del (Siempre y cuando esten definidas). Por ejemplo, el siguiente código elimina la variable edad y nombre:

```python
del edad, nombre
print(altura)
```

Cuidado! Que si intento eliminar la variable edad, sin crearla de nuevo, dará error, ya que la misma no existe. Este codigo resuelve este problema:


```python
if "edad" in globals():
    del edad
    print("puedo borrar edad")
else:
    print("no puedo borrar")
```

*Nota 3: globals es un diccionario que cuenta con todas las variables globales que se utilizan en el programa actual.*

```python

# ==========================================
# EXPERIMENTANDO CON VARIABLES Y DEL
# ==========================================

# 1. Creamos nuestras variables
edad = 20
nombre = "Anthony"
altura = 1.75
ciudad = "Montevideo"

print("===== VARIABLES CREADAS =====")

print("Edad:", edad)
print("Nombre:", nombre)
print("Altura:", altura)
print("Ciudad:", ciudad)


# 2. Comprobamos si existen
print("\n===== COMPROBAR VARIABLES =====")

if "edad" in globals():
    print("La variable edad existe.")
else:
    print("La variable edad no existe.")

if "nombre" in globals():
    print("La variable nombre existe.")
else:
    print("La variable nombre no existe.")


# 3. Eliminamos algunas variables
print("\n===== ELIMINAR VARIABLES =====")

if "edad" in globals():
    del edad
    print("La variable edad fue eliminada.")
else:
    print("No se puede eliminar edad porque no existe.")

if "ciudad" in globals():
    del ciudad
    print("La variable ciudad fue eliminada.")
else:
    print("La variable ciudad no existe.")


# 4. Comprobamos si siguen existiendo
print("\n===== DESPUES DE ELIMINAR =====")

if "edad" not in globals():
    print("Edad ya no existe.")

if "ciudad" not in globals():
    print("Ciudad ya no existe.")

if "nombre" in globals():
    print("Nombre sigue existiendo:", nombre)

if "altura" in globals():
    print("Altura sigue existiendo:", altura)


# 5. Volvemos a crear las variables
print("\n===== RECUPERAR VARIABLES =====")

edad = 25
ciudad = "Punta del Este"

print("Nueva edad:", edad)
print("Nueva ciudad:", ciudad)

print("\n===== FIN DEL PROGRAMA =====")
```

# Tipos de datos simples. Enteros, reales, booleanos y caracteres.

En programación, los datos son la información que se manipula. Los lenguajes de programación trabajan con diferentes tipos de datos para poder manipular la información. En este artículo nos enfocaremos en los tipos de datos simples en Python, que son los más básicos y fundamentales para cualquier programa.

Python es un lenguaje de programación orientado a objetos, lo que significa que cada valor en Python es un objeto. Un objeto es una estructura de datos que puede tener atributos (variables) y métodos (funciones). Los objetos en Python tienen un tipo que determina qué operaciones se pueden realizar con ellos.

## Tipos de datos simples

Los tipos de datos simples en Python se dividen en cuatro categorías: enteros, reales, booleanos y caracteres. En Python también existe un tipo de dato complejo, que nos permite trabajar con números complejos.

### Enteros (int)

Los enteros son números sin parte decimal. En Python, los enteros se representan con el tipo de datos int. Los enteros se pueden representar en notación decimal, binaria, octal y hexadecimal.

Por ejemplo, para definir una variable entera y mostrar su valor y tipo, podemos hacer lo siguiente:

```python
numero_entero = 10
print("Número entero:", numero_entero, "Tipo:", type(numero_entero))
```

```python
a = 3
b = 5
c = 10
res = 0

res=a+b
print(res)
res=a-b
print(res)
res=a*b
print(res)
res=a/b
print(res)
res=c/a
print(res)
res=a**b
print(res)
```

```python

# ==========================================
# TIPOS DE DATOS SIMPLES: ENTEROS (INT)
# ==========================================

# 1. Creamos variables enteras
a = 15
b = 4
c = 10

print("===== NUMEROS ENTEROS =====")

print("Primer numero:", a, "Tipo:", type(a))
print("Segundo numero:", b, "Tipo:", type(b))
print("Tercer numero:", c, "Tipo:", type(c))


# 2. Operaciones matematicas
print("\n===== OPERACIONES BASICAS =====")

suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

print("Suma:", suma)
print("Resta:", resta)
print("Multiplicacion:", multiplicacion)
print("Division:", division)


# 3. Operaciones adicionales
print("\n===== OTRAS OPERACIONES =====")

division_entera = a // b
resto = a % b
potencia = a ** b

print("Division entera:", division_entera)
print("Resto de la division:", resto)
print("Potencia:", potencia)


# 4. Comparamos numeros
print("\n===== COMPARACIONES =====")

print("¿A es mayor que B?", a > b)
print("¿A es menor que B?", a < b)
print("¿A es igual a B?", a == b)
print("¿A es diferente de B?", a != b)


# 5. Comprobamos si son pares o impares
print("\n===== PARES E IMPARES =====")

if a % 2 == 0:
    print(a, "es un numero par.")
else:
    print(a, "es un numero impar.")

if b % 2 == 0:
    print(b, "es un numero par.")
else:
    print(b, "es un numero impar.")


# 6. Convertimos numeros
print("\n===== CONVERSION DE DATOS =====")

numero_decimal = float(a)
numero_texto = str(b)

print("A como decimal:", numero_decimal, type(numero_decimal))
print("B como texto:", numero_texto, type(numero_texto))


# 7. Representaciones de enteros
print("\n===== REPRESENTACIONES =====")

print("Decimal:", c)
print("Binario:", bin(c))
print("Octal:", oct(c))
print("Hexadecimal:", hex(c))

print("\n===== FIN DEL PROGRAMA =====")
```

### Reales (float)

Los números reales son aquellos que tienen una parte decimal. En Python, los números reales se representan con el tipo de datos float.

Por ejemplo, para definir una variable real y mostrar su valor y tipo, podemos hacer lo siguiente:

```python
numero_real = 3.1416
print("Número real:", numero_real, "Tipo:", type(numero_real))
```

```python

# ==========================================
# TIPOS DE DATOS SIMPLES: REALES (FLOAT)
# ==========================================

# 1. Creamos variables decimales
numero_real = 3.1416
precio = 125.50
altura = 1.68
temperatura = 24.5

print("===== NUMEROS REALES =====")

print("Numero real:", numero_real, "Tipo:", type(numero_real))
print("Precio:", precio, "Tipo:", type(precio))
print("Altura:", altura, "Tipo:", type(altura))
print("Temperatura:", temperatura, "Tipo:", type(temperatura))


# 2. Operaciones con decimales
print("\n===== OPERACIONES MATEMATICAS =====")

a = 12.5
b = 3.2

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicacion:", a * b)
print("Division:", a / b)
print("Potencia:", a ** 2)


# 3. Redondeamos los resultados
print("\n===== REDONDEO DE DECIMALES =====")

division = a / b

print("Resultado original:", division)
print("Redondeado a 2 decimales:", round(division, 2))
print("Redondeado a 1 decimal:", round(division, 1))
print("Redondeado a entero:", round(division))


# 4. Calculamos descuentos
print("\n===== CALCULADORA DE DESCUENTOS =====")

precio_producto = 850.75
porcentaje_descuento = 15.0

descuento = precio_producto * porcentaje_descuento / 100
precio_final = precio_producto - descuento

print("Precio original: $", precio_producto)
print("Descuento:", porcentaje_descuento, "%")
print("Dinero descontado: $", round(descuento, 2))
print("Precio final: $", round(precio_final, 2))


# 5. Convertimos temperaturas
print("\n===== CONVERSION DE TEMPERATURA =====")

celsius = 25.5
fahrenheit = (celsius * 9 / 5) + 32

print("Temperatura en Celsius:", celsius)
print("Temperatura en Fahrenheit:", fahrenheit)


# 6. Convertimos tipos de datos
print("\n===== CONVERSION DE TIPOS =====")

numero_entero = int(numero_real)
numero_texto = str(numero_real)

print("Decimal original:", numero_real, type(numero_real))
print("Convertido a entero:", numero_entero, type(numero_entero))
print("Convertido a texto:", numero_texto, type(numero_texto))


# 7. Calculamos el area de un circulo
print("\n===== AREA DE UN CIRCULO =====")

pi = 3.1416
radio = 5.5

area = pi * radio ** 2

print("Radio del circulo:", radio)
print("Area del circulo:", round(area, 2))

print("\n===== FIN DEL PROGRAMA =====")
```

### Booleanos (bool)

Los booleanos son tipos de datos que solo pueden tomar dos valores: True (verdadero) o False (falso). En Python, los booleanos se representan con el tipo de datos bool.

Por ejemplo, para definir una variable booleana y mostrar su valor y tipo, podemos hacer lo siguiente:

```python
verdadero = True
print("Booleano:", verdadero, "Tipo:", type(verdadero))
```

Los valores booleanos tambien pueden ser utilizados

```python
A = True
B = False
C = True
D = A and B
print(D)
```

```python

# ==========================================
# TIPOS DE DATOS SIMPLES: BOOLEANOS (BOOL)
# ==========================================

# 1. Creamos variables booleanas
verdadero = True
falso = False
es_estudiante = True
tiene_tarea = False

print("===== VARIABLES BOOLEANAS =====")

print("Verdadero:", verdadero, "Tipo:", type(verdadero))
print("Falso:", falso, "Tipo:", type(falso))
print("Es estudiante:", es_estudiante)
print("Tiene tarea:", tiene_tarea)


# 2. Comparaciones que devuelven booleanos
print("\n===== COMPARACIONES =====")

edad = 20
nota = 8

print("¿Es mayor de edad?", edad >= 18)
print("¿La nota es mayor que 6?", nota > 6)
print("¿La edad es igual a 20?", edad == 20)
print("¿La nota es diferente de 10?", nota != 10)


# 3. Operadores logicos
print("\n===== OPERADORES LOGICOS =====")

print("AND:", es_estudiante and tiene_tarea)
print("OR:", es_estudiante or tiene_tarea)
print("NOT:", not es_estudiante)


# 4. Sistema para comprobar si aprobo
print("\n===== RESULTADO DEL EXAMEN =====")

nombre = "Anthony"
calificacion = 8
nota_minima = 6

aprobo = calificacion >= nota_minima

print("Estudiante:", nombre)
print("Calificacion:", calificacion)
print("¿Aprobo?", aprobo)

if aprobo:
    print("¡Felicitaciones! Aprobaste el examen.")
else:
    print("Debes seguir estudiando para aprobar.")


# 5. Sistema de acceso
print("\n===== CONTROL DE ACCESO =====")

tiene_usuario = True
tiene_contrasena = True

puede_ingresar = tiene_usuario and tiene_contrasena

print("¿Puede ingresar?", puede_ingresar)

if puede_ingresar:
    print("Acceso permitido.")
else:
    print("Acceso denegado.")


# 6. Conversion de datos a booleanos
print("\n===== CONVERSION A BOOLEANOS =====")

print("Numero 1:", bool(1))
print("Numero 0:", bool(0))
print("Texto Hola:", bool("Hola"))
print("Texto vacio:", bool(""))

print("\n===== FIN DEL PROGRAMA =====")
```

*Nota 5: Como operadores logicos solo existen 3, el and, or, y el not.
Faltando: el condicional, bicondicional*

### Caracteres (str)

Los caracteres son un tipo de dato que representa un solo carácter. En Python, los caracteres se representan con el tipo de datos str.

Por ejemplo, para definir una variable de tipo carácter y mostrar su valor y tipo, podemos hacer lo siguiente:

```python
caracter = 'a'
print("Carácter:", caracter, "Tipo:", type(caracter))
```

```python

# ==========================================
# TIPOS DE DATOS: CARACTERES Y TEXTO (STR)
# ==========================================

# 1. Creamos variables de texto
caracter = 'a'
nombre = "Anthony"
apellido = "Perez"
ciudad = "Montevideo"
mensaje = "Estoy aprendiendo Python"

print("===== VARIABLES DE TEXTO =====")

print("Caracter:", caracter, "Tipo:", type(caracter))
print("Nombre:", nombre, "Tipo:", type(nombre))
print("Apellido:", apellido)
print("Ciudad:", ciudad)
print("Mensaje:", mensaje)


# 2. Unimos cadenas de texto
print("\n===== UNIR TEXTOS =====")

nombre_completo = nombre + " " + apellido

print("Nombre completo:", nombre_completo)
print("Saludo:", "Hola, " + nombre + "!")


# 3. Transformamos el texto
print("\n===== TRANSFORMACIONES =====")

print("Mayusculas:", nombre.upper())
print("Minusculas:", nombre.lower())
print("Primera letra mayuscula:", mensaje.capitalize())
print("Titulo:", mensaje.title())


# 4. Contamos caracteres
print("\n===== LONGITUD DEL TEXTO =====")

print("Cantidad de letras del nombre:", len(nombre))
print("Cantidad de caracteres del mensaje:", len(mensaje))


# 5. Accedemos a caracteres individuales
print("\n===== CARACTERES INDIVIDUALES =====")

print("Primera letra:", nombre[0])
print("Ultima letra:", nombre[-1])
print("Primeras tres letras:", nombre[:3])


# 6. Reemplazamos palabras
print("\n===== REEMPLAZAR TEXTO =====")

nuevo_mensaje = mensaje.replace("Python", "programacion")

print("Mensaje original:", mensaje)
print("Mensaje modificado:", nuevo_mensaje)


# 7. Comprobamos si un texto contiene otro
print("\n===== BUSCAR PALABRAS =====")

print("¿Contiene Python?", "Python" in mensaje)
print("¿Contiene Java?", "Java" in mensaje)


# 8. Repetimos caracteres
print("\n===== REPETIR CARACTERES =====")

print("*" * 25)
print("BIENVENIDO A PYTHON")
print("*" * 25)

print("\n===== FIN DEL PROGRAMA =====")
```

### Complejos (complex)

Los números complejos son aquellos que tienen una parte real y una parte imaginaria. En Python, los números complejos se representan con el tipo de datos complex.

Por ejemplo, para definir una variable compleja y mostrar su valor y tipo, podemos hacer lo siguiente:

```python
numero_complejo = 2+3j
print("Número complejo:", numero_complejo, "Tipo:", type(numero_complejo))
```

```python

# ==========================================
# TIPOS DE DATOS: NUMEROS COMPLEJOS
# ==========================================

# 1. Creamos numeros complejos
numero_complejo = 2 + 3j
otro_complejo = 4 + 2j

print("===== NUMEROS COMPLEJOS =====")

print("Primer numero:", numero_complejo)
print("Tipo:", type(numero_complejo))

print("Segundo numero:", otro_complejo)
print("Tipo:", type(otro_complejo))


# 2. Identificamos sus partes
print("\n===== PARTES DEL NUMERO =====")

print("Parte real:", numero_complejo.real)
print("Parte imaginaria:", numero_complejo.imag)


# 3. Operaciones matematicas
print("\n===== OPERACIONES =====")

suma = numero_complejo + otro_complejo
resta = numero_complejo - otro_complejo
multiplicacion = numero_complejo * otro_complejo
division = numero_complejo / otro_complejo

print("Suma:", suma)
print("Resta:", resta)
print("Multiplicacion:", multiplicacion)
print("Division:", division)


# 4. Calculamos potencias
print("\n===== POTENCIAS =====")

cuadrado = numero_complejo ** 2
cubo = numero_complejo ** 3

print("Elevado al cuadrado:", cuadrado)
print("Elevado al cubo:", cubo)


# 5. Obtenemos el conjugado
print("\n===== CONJUGADO =====")

conjugado = numero_complejo.conjugate()

print("Numero original:", numero_complejo)
print("Conjugado:", conjugado)


# 6. Calculamos la magnitud
print("\n===== MAGNITUD =====")

magnitud = abs(numero_complejo)

print("Magnitud:", round(magnitud, 2))


# 7. Creamos otro numero complejo
print("\n===== CREAR COMPLEJOS =====")

nuevo_complejo = complex(5, 7)

print("Nuevo numero:", nuevo_complejo)
print("Parte real:", nuevo_complejo.real)
print("Parte imaginaria:", nuevo_complejo.imag)

print("\n===== FIN DEL PROGRAMA =====")
```

# Operadores logicos y de comparacion

Los operadores lógicos y de comparación son fundamentales en la programación, ya que permiten evaluar condiciones y realizar acciones basadas en el resultado de esas condiciones. En Python, tenemos operadores de comparación para comparar valores y operadores lógicos para combinar condiciones.

## Operadores de comparacion
Los operadores de comparación son utilizados para comparar dos valores y devuelven un valor booleano, es decir, True o False. Los operadores de comparación en Python incluyen:

* Igual (==)
* Diferente (!=)
* Mayor que (>)
* Menor que (<)
* Mayor o igual que (>=)
* Menor o igual que (<=)

Ejemplo de uso de operadores de comparación:

```python
a = 5
b = 7

print("a == b:", a == b) # Falso, porque 5 no es igual a 7
print("a != b:", a != b) # Verdadero, porque 5 es diferente de 7
print("a > b:", a > b)   # Falso, porque 5 no es mayor que 7
print("a < b:", a < b)   # Verdadero, porque 5 es menor que 7
print("a >= b:", a >= b) # Falso, porque 5 no es mayor o igual que 7
print("a <= b:", a <= b) # Verdadero, porque 5 es menor o igual que 7
```

## Operadores logicos
Los operadores lógicos son utilizados para combinar condiciones. Los operadores lógicos en Python incluyen:

* and: Devuelve True si ambas condiciones son verdaderas, de lo contrario devuelve False.
* or: Devuelve True si al menos una de las condiciones es verdadera, de lo contrario devuelve False.
* not: Devuelve True si la condición es falsa, y False si la condición es verdadera.

Ejemplo de uso de operadores lógicos:

```python
x = 5
y = 10
z = 15

# Comprobar si x es menor que y y si y es menor que z
condicion1 = x < y and y < z
print("Condición 1 (x < y y y < z):", condicion1) # Verdadero

# Comprobar si x es igual a y o si y es igual a z
condicion2 = x == y or y == z
print("Condición 2 (x == y o y == z):", condicion2) # Falso

# Negar la condición 1
condicion3 = not condicion1
print("Condición 3 (negación de la condición 1):", condicion3) # Falso
```

```python

# ==========================================
# OPERADORES LOGICOS EN PYTHON
# ==========================================

# 1. Creamos nuestras variables
x = 5
y = 10
z = 15

print("===== OPERADORES LOGICOS =====")

print("X:", x)
print("Y:", y)
print("Z:", z)

# 2. Operador AND
print("\n===== OPERADOR AND =====")

condicion1 = x < y and y < z

print("¿X es menor que Y y Y menor que Z?", condicion1)

# 3. Operador OR
print("\n===== OPERADOR OR =====")

condicion2 = x == y or y == z

print("¿X es igual a Y o Y igual a Z?", condicion2)

# 4. Operador NOT
print("\n===== OPERADOR NOT =====")

condicion3 = not condicion1

print("Negacion de la primera condicion:", condicion3)


# 5. Ejemplo: entrada a una fiesta
print("\n===== ENTRADA A UNA FIESTA =====")

edad = 20
tiene_invitacion = True

puede_entrar = edad >= 18 and tiene_invitacion

print("Edad:", edad)
print("Tiene invitacion:", tiene_invitacion)
print("¿Puede entrar?", puede_entrar)

if puede_entrar:
    print("¡Bienvenido a la fiesta!")
else:
    print("No puedes ingresar.")


# 6. Ejemplo: aprobar una materia
print("\n===== APROBACION DE MATERIA =====")

nota = 8
entrego_trabajo = True

aprobo = nota >= 6 and entrego_trabajo

print("Calificacion:", nota)
print("Entrego el trabajo:", entrego_trabajo)

if aprobo:
    print("¡Felicitaciones! Aprobaste.")
else:
    print("Todavia no cumples los requisitos.")


# 7. Ejemplo: descuentos especiales
print("\n===== DESCUENTOS ESPECIALES =====")

es_estudiante = True
es_docente = False

tiene_descuento = es_estudiante or es_docente

print("¿Tiene descuento?", tiene_descuento)

if tiene_descuento:
    print("¡Tienes un descuento especial!")
else:
    print("No tienes descuento.")


# 8. Ejemplo: sistema de acceso
print("\n===== SISTEMA DE ACCESO =====")

usuario_activo = True
cuenta_bloqueada = False

acceso = usuario_activo and not cuenta_bloqueada

print("Usuario activo:", usuario_activo)
print("Cuenta bloqueada:", cuenta_bloqueada)
print("¿Acceso permitido?", acceso)

if acceso:
    print("Acceso autorizado.")
else:
    print("Acceso denegado.")


# 9. Combinamos los tres operadores
print("\n===== CONDICIONES COMBINADAS =====")

edad = 19
tiene_permiso = False
es_invitado = True

resultado = (edad >= 18 or tiene_permiso) and es_invitado

print("¿Cumple las condiciones?", resultado)

print("\n===== FIN DEL PROGRAMA =====")
```

# Comparaciones encadenadas

Una característica interesante de Python es la posibilidad de realizar comparaciones encadenadas, lo que permite escribir condiciones más compactas y legibles. Veamos un ejemplo de cómo podemos utilizar comparaciones encadenadas y luego plantearemos un desafío para poner en práctica este concepto.

En Python, se pueden encadenar comparaciones en una sola expresión. Por ejemplo, en lugar de escribir x < y and y < z, puedes escribir x < y < z. Esto es posible gracias a la evaluación de cortocircuito de Python, lo que significa que si la primera comparación devuelve False, el resto de las comparaciones no se evaluarán.

Ejemplo de comparaciones encadenadas:

```python
a = 5
b = 10
c = 15

# Comprobar si a es menor que b y si b es menor que c
condicion_encadenada = a < b < c
print("Comparación encadenada (a < b < c):", condicion_encadenada) # Verdadero
```

# Desafío 3: Convertidor de temperaturas
Realizar conversiones de una temperatura dada en Celsius a Fahrenheit y Kelvin utilizando operaciones directas sobre las variables. Los cálculos se realizarán directamente al asignar los valores a las variables.

```python

# ==========================================
# DESAFIO 3: CONVERTIDOR DE TEMPERATURAS
# ==========================================

# Temperatura inicial en Celsius
celsius = 25.5

print("===== CONVERTIDOR DE TEMPERATURAS =====")

# Conversiones directas
fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print("Temperatura en Celsius:", celsius, "°C")
print("Temperatura en Fahrenheit:",
      round(fahrenheit, 2), "°F")
print("Temperatura en Kelvin:",
      round(kelvin, 2), "K")

# Clasificamos la temperatura
print("\n===== ANALISIS DE TEMPERATURA =====")

if celsius < -273.15:
    print("Temperatura inferior al cero absoluto.")

elif celsius < 0:
    print("Temperatura bajo cero.")

elif celsius == 0:
    print("Punto de congelacion del agua.")

elif 0 < celsius < 30:
    print("Temperatura templada o calida.")

else:
    print("Temperatura elevada.")

print("\n===== FIN DEL DESAFIO 3 =====")
```

# Desafío 4: Verificar múltiplos de varios números
Objetivo: Determinar si un número dado es múltiplo de 2, 3, 5, 7, 9, 10, y 11 utilizando comparaciones y operadores lógicos.

*Descripción:*
Crea un programa que, dado un número, verifique si es múltiplo de los números mencionados. Para cada uno de estos números, el programa debe indicar si el número dado es o no un múltiplo de estos.

_Ejemplo de lo que se debe lograr:_
Supongamos que el número dado es 30. El programa deberá determinar:

¿Es 30 múltiplo de 2? Sí

¿Es 30 múltiplo de 3? Sí

¿Es 30 múltiplo de 5? Sí

¿Es 30 múltiplo de 7? No

¿Es 30 múltiplo de 9? No

¿Es 30 múltiplo de 10? Sí

¿Es 30 múltiplo de 11? No


```python

# ==========================================
# DESAFIO 4: VERIFICAR MULTIPLOS
# ==========================================

# Numero que vamos a analizar
numero = 30

print("===== VERIFICADOR DE MULTIPLOS =====")
print("Numero elegido:", numero)

# Comprobamos cada multiplo
multiplo_2 = numero % 2 == 0
multiplo_3 = numero % 3 == 0
multiplo_5 = numero % 5 == 0
multiplo_7 = numero % 7 == 0
multiplo_9 = numero % 9 == 0
multiplo_10 = numero % 10 == 0
multiplo_11 = numero % 11 == 0

print("\n===== RESULTADOS =====")

print("¿Es multiplo de 2?",
      "Si" if multiplo_2 else "No")

print("¿Es multiplo de 3?",
      "Si" if multiplo_3 else "No")

print("¿Es multiplo de 5?",
      "Si" if multiplo_5 else "No")

print("¿Es multiplo de 7?",
      "Si" if multiplo_7 else "No")

print("¿Es multiplo de 9?",
      "Si" if multiplo_9 else "No")

print("¿Es multiplo de 10?",
      "Si" if multiplo_10 else "No")

print("¿Es multiplo de 11?",
      "Si" if multiplo_11 else "No")


# Combinamos condiciones con AND
print("\n===== MULTIPLOS COMBINADOS =====")

if multiplo_2 and multiplo_3:
    print("Es multiplo de 2 y de 3.")

if multiplo_3 and multiplo_5:
    print("Es multiplo de 3 y de 5.")

# Utilizamos OR
if multiplo_7 or multiplo_11:
    print("Es multiplo de 7 o de 11.")
else:
    print("No es multiplo de 7 ni de 11.")

# Contamos cuantos divisores de la lista cumple
cantidad = sum([
    multiplo_2, multiplo_3, multiplo_5,
    multiplo_7, multiplo_9, multiplo_10,
    multiplo_11
])

print("\nCantidad de divisores encontrados:", cantidad)

print("\n===== FIN DEL DESAFIO 4 =====")
```
