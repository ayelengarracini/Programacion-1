Desafío 14: Contador de palabras

Escriba un programa que permita al usuario ingresar un texto y luego cuente y muestre la cantidad de palabras que contiene el texto. El programa debe manejar la entrada de datos y formatear la salida adecuadamente.

```python

# ==========================================
# DESAFIO: CONTADOR DE PALABRAS
# ==========================================

print("===== CONTADOR DE PALABRAS =====")

# 1. Definimos un texto de ejemplo
texto = "Hola, soy Ayelén y estoy aprendiendo Python en Visual Studio Code."

# 2. Separamos el texto en palabras
palabras = texto.split()

# 3. Contamos las palabras
cantidad_palabras = len(palabras)

# 4. Contamos los caracteres
cantidad_caracteres = len(texto)

# 5. Contamos los caracteres sin espacios
caracteres_sin_espacios = len("".join(texto.split()))

# 6. Mostramos los resultados
print("\n===== TEXTO INGRESADO =====")
print(texto)

print("\n===== RESULTADOS =====")
print(f"Cantidad de palabras: {cantidad_palabras}")
print(f"Cantidad de caracteres: {cantidad_caracteres}")
print(f"Caracteres sin espacios: {caracteres_sin_espacios}")

# 7. Mostramos las palabras individualmente
print("\n===== LISTA DE PALABRAS =====")

for numero, palabra in enumerate(palabras, start=1):
    print(f"{numero}. {palabra}")

print("\n===== FIN DEL PROGRAMA =====")
```
