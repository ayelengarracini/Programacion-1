# Colecciones avanzadas en Python: Tuplas, Diccionarios y Conjuntos
En este cuaderno, exploraremos colecciones avanzadas en Python: tuplas, diccionarios y conjuntos. Aunque ya hemos trabajado con listas y árboles en secciones anteriores, ahora enfrentaremos problemas que requieren estructuras de datos más especializadas para ser resueltos de manera eficiente.

## Tuplas
Problema: Registro de Coordenadas Inmutables
Imagina que estás desarrollando un sistema de seguimiento de satélites donde necesitas almacenar las coordenadas (latitud, longitud) de cada satélite en un momento dado. Estas coordenadas no deben cambiar una vez registradas para mantener un historial preciso. ¿Cómo puedes asegurar que estos datos permanezcan inmutables y se almacenen eficientemente?

### Limitaciones de las Listas
Las listas en Python son mutables, lo que significa que sus elementos pueden cambiarse después de su creación. Esto es útil en muchos casos, pero en situaciones donde la inmutabilidad es clave, como en nuestro problema, las listas no son la mejor opción.

### Definición y Uso de Tuplas
Las tuplas en Python son colecciones ordenadas e inmutables de elementos. Se definen utilizando paréntesis () y son ideales para almacenar datos que no deben cambiar a lo largo del programa.

### Creación de una Tupla

```python
# Tupla de coordenadas
coordenada = (-34.46262, -57.83976)
print(coordenada)
```

```python
# Tuplas de coordenadas
coordenada1 = (-34.46262, -57.83976)
coordenada2 = (-33.50000, -56.90000)

print("COORDENADAS REGISTRADAS")

print("Satélite 1:", coordenada1)
print("Satélite 2:", coordenada2)

print("\nCantidad de valores en cada tupla:")
print("Satélite 1:", len(coordenada1))
print("Satélite 2:", len(coordenada2))
```

### Características de las Tuplas
* Inmutabilidad: Una vez creada, no puedes modificar sus elementos.
* Ordenadas: Mantienen el orden de inserción de los elementos.
* Acceso por Índice: Puedes acceder a sus elementos utilizando índices, igual que en las listas.

### Acceso a Elementos

```python
latitud = coordenada[0]
longitud = coordenada[1]
print(f"Latitud: {latitud}, Longitud: {longitud}")
```

```python
# Accedemos a las coordenadas del primer satélite
latitud1 = coordenada1[0]
longitud1 = coordenada1[1]

# Accedemos a las coordenadas del segundo satélite
latitud2 = coordenada2[0]
longitud2 = coordenada2[1]

print("COORDENADAS DEL SATÉLITE 1")
print("Latitud:", latitud1)
print("Longitud:", longitud1)

print("\nCOORDENADAS DEL SATÉLITE 2")
print("Latitud:", latitud2)
print("Longitud:", longitud2)
```

### Desempaquetado de Tuplas
El desempaquetado permite asignar los valores de una tupla a variables individuales de manera concisa.

```python
# Desempaquetado
latitud, longitud = coordenada
print(f"Latitud: {latitud}, Longitud: {longitud}")
```

```python
# Desempaquetamos las coordenadas
latitud1, longitud1 = coordenada1
latitud2, longitud2 = coordenada2

print("DESEMPAQUETADO DE TUPLAS")

print("\nSatélite 1")
print("Latitud:", latitud1)
print("Longitud:", longitud1)

print("\nSatélite 2")
print("Latitud:", latitud2)
print("Longitud:", longitud2)

# Comparamos las coordenadas
print("\nCOMPARACIÓN")

if latitud1 == latitud2:
    print("Los satélites tienen la misma latitud.")
else:
    print("Los satélites tienen latitudes diferentes.")
```

### Métodos y Operaciones con Tuplas
Aunque las tuplas son inmutables, puedes realizar ciertas operaciones:

* Concatenación: Combinar dos tuplas.

```python
tupla1 = (1, 2, 3)
tupla2 = (4, 5, 6)
tupla_concatenada = tupla1 + tupla2
print(tupla_concatenada)
```

```python
tupla1 = (1, 2, 3, 4)
tupla2 = (5, 6, 7, 8)

tupla_concatenada = tupla1 + tupla2

print("Primera tupla:", tupla1)
print("Segunda tupla:", tupla2)
print("Tupla concatenada:", tupla_concatenada)
```

* Repetición: Repetir los elementos de una tupla.

```python
tupla_repetida = tupla1 * 3
print(tupla_repetida)
```

```python
tupla1 = (1, 2, 3)

tupla_repetida = tupla1 * 4

print("Tupla original:", tupla1)
print("Tupla repetida:", tupla_repetida)
```

### Métodos disponibles:

* count(): Cuenta las ocurrencias de un elemento.
* index(): Encuentra el índice de un elemento.

```python
tupla = (1, 2, 3, 2, 4, 2, 5, 3, 3)

conteo_2 = tupla.count(2)
conteo_3 = tupla.count(3)

print("Tupla:", tupla)
print("El número 2 aparece:", conteo_2, "veces")
print("El número 3 aparece:", conteo_3, "veces")
```

```python
tupla = (1, 2, 3, 2, 4, 2, 5, 3, 3)

indice_2 = tupla.index(2)
indice_5 = tupla.index(5)

print("Tupla:", tupla)
print("Primera posición del número 2:", indice_2)
print("Primera posición del número 5:", indice_5)
```

```python
tupla = (1, 2, 3, 2, 2)
print(tupla.count(2))     # Salida: 3
print(tupla.index(3))     # Salida: 2
```

## Diccionarios
Problema: Traducción de Palabras en Múltiples Idiomas
Estás creando una aplicación de traducción que necesita mapear palabras en inglés a sus equivalentes en varios idiomas. Las relaciones entre palabras no son simplemente lineales y requieren un método eficiente para acceder a las traducciones utilizando la palabra original como clave. ¿Cómo podrías implementar esto de manera eficiente?

Limitaciones de las Listas y Tuplas
Las listas y tuplas no son adecuadas para este problema porque requieren búsquedas lineales para encontrar elementos, lo cual no es eficiente para grandes conjuntos de datos. Además, no proporcionan una asociación directa entre una clave y un valor.

Definición y Uso de Diccionarios
Los diccionarios en Python son colecciones desordenadas de pares clave-valor. Se definen utilizando llaves {} y permiten un acceso rápido a los valores a través de sus claves únicas.

### Creación de un Diccionario

```python
# Diccionario de traducciones
traducciones = {
    "hello": {"es": "hola", "fr": "bonjour", "pr": "olá"},
    "world": {"es": "mundo", "fr": "monde", "pr": "mundo"}
}
```

En este diccionario:

La clave principal es la palabra en inglés.
El valor es otro diccionario que contiene las traducciones en diferentes idiomas:
* "es": Español
* "fr": Francés
* "pr": Portugués

```python
traducciones = {
    "hello": {"es": "hola", "fr": "bonjour", "pt": "olá"},
    "world": {"es": "mundo", "fr": "monde", "pt": "mundo"},
    "goodbye": {"es": "adiós", "fr": "au revoir", "pt": "adeus"}
}

print("Diccionario de traducciones:")
print(traducciones)
```

### Acceso a valores
Para acceder a la traducción de una palabra en un idioma específico, utilizamos ambas claves:

```python
# Obtener la traducción de "hello" en español
print(traducciones["hello"]["es"])  # Salida: hola

# Obtener la traducción de "world" en portugués
print(traducciones["world"]["pr"])  # Salida: mundo
```

```python
print("Hello en español:", traducciones["hello"]["es"])

print("World en francés:", traducciones["world"]["fr"])

print("Goodbye en portugués:", traducciones["goodbye"]["pt"])
```

### Operaciones básicas con Diccionarios
* Agregar Nuevos Elementos
  
  Podemos agregar nuevas palabras y sus traducciones al diccionario:

```python
# Agregar una nueva palabra y sus traducciones
traducciones["goodbye"] = {"es": "adiós", "fr": "au revoir", "pr": "adeus"}

print(traducciones["goodbye"]["pr"])  # Salida: adeus
```

```python
traducciones["thanks"] = {
    "es": "gracias",
    "fr": "merci",
    "pt": "obrigado"
}

print("Nueva palabra agregada:")
print(traducciones["thanks"])

print("Traducción al francés:", traducciones["thanks"]["fr"])
```

* Modificar valores existentes
  
Si necesitamos actualizar una traducción específica, simplemente asignamos un nuevo valor:

```python
# Actualizar la traducción en portugués de "world"
traducciones["world"]["pr"] = "mundo"

print(traducciones["world"]["pr"])  # Salida: mundo
```

```python
print("Traducción anterior:", traducciones["goodbye"]["fr"])

traducciones["goodbye"]["fr"] = "salut"

print("Traducción modificada:", traducciones["goodbye"]["fr"])
```

* Eliminar Elementos

Podemos eliminar una palabra completa o una traducción específica:

```python
# Eliminar una palabra completa
del traducciones["goodbye"]

# Intentar acceder a "goodbye" genera un KeyError
# print(traducciones["goodbye"])  # Esto causaría un error

# Eliminar la traducción al francés de "hello"
del traducciones["hello"]["fr"]

print(traducciones["hello"])  # Salida: {'es': 'hola', 'pr': 'olá'}
```

```python
# Eliminar una palabra completa
del traducciones["thanks"]

print("Diccionario después de eliminar thanks:")
print(traducciones)

# Eliminar solamente una traducción
del traducciones["hello"]["fr"]

print("Traducciones restantes de hello:")
print(traducciones["hello"])
```

### Iteración y métodos útiles
* Iterar sobre Claves y Valores

Podemos recorrer el diccionario para procesar sus elementos:

```python
for palabra, traduccion in traducciones.items():
    print(f"{palabra}: {traduccion}")
```

```python
for palabra, traduccion in traducciones.items():
    print("Palabra en inglés:", palabra)
    print("Traducciones:", traduccion)
    print("--------------------")
```

* Métodos Útiles

keys(): Devuelve una vista de las claves.

```python
print(traducciones.keys())
# Salida: dict_keys(['hello', 'world'])
```

```python
print("Palabras disponibles:")
print(traducciones.keys())

print("\nLista de palabras:")

for palabra in traducciones.keys():
    print(palabra)
```

values(): Devuelve una vista de los valores.

```python
print(traducciones.values())
# Salida: dict_values([{'es': 'hola', 'pr': 'olá'}, {'es': 'mundo', 'fr': 'monde', 'pr': 'mundo'}])
```

```python
print("Traducciones disponibles:")
print(traducciones.values())

print("\nTraducciones por palabra:")

for traduccion in traducciones.values():
    print(traduccion)
```

items(): Devuelve una vista de los pares clave-valor.

```python
print(traducciones.items())
# Salida: dict_items([('hello', {'es': 'hola', 'pr': 'olá'}), ('world', {'es': 'mundo', 'fr': 'monde', 'pr': 'mundo'})])
```

```python
print("Contenido del diccionario:")
print(traducciones.items())

print("\nPalabras y traducciones:")

for palabra, traduccion in traducciones.items():
    print(palabra, ":", traduccion)
```

### Verificar la existencia de una clave
Podemos verificar si una palabra o una traducción específica existe en el diccionario:

```python
# Verificar si "hello" está en el diccionario
print("hello" in traducciones)  # Salida: True

# Verificar si existe traducción al francés para "hello"
print("fr" in traducciones["hello"])  # Salida: False
```

```python
print("¿Existe hello?", "hello" in traducciones)

print("¿Existe goodbye?", "goodbye" in traducciones)

print("¿Existe la traducción al francés de hello?",
      "fr" in traducciones["hello"])
```

### Añadir una nueva traducción a una palabra existente
Si queremos agregar una nueva traducción a una palabra existente:

```python
# Añadir traducción al francés para "hello"
traducciones["hello"]["fr"] = "bonjour"

print(traducciones["hello"])
# Salida: {'es': 'hola', 'pr': 'olá', 'fr': 'bonjour'}
```

```python
print("Traducciones originales:")
print(traducciones["hello"])

traducciones["hello"]["fr"] = "bonjour"

print("\nTraducciones actualizadas:")
print(traducciones["hello"])
```

### Aplicación práctica: Traductor simple
Podemos utilizar el diccionario para crear una función que traduzca palabras:

```python
def traducir(palabra_ingles, idioma):
    try:
        return traducciones[palabra_ingles][idioma]
    except KeyError:
        return "Traducción no encontrada"

# Ejemplos de uso
print(traducir("hello", "es"))  # Salida: hola
print(traducir("world", "fr"))  # Salida: monde
print(traducir("goodbye", "pr"))  # Salida: Traducción no encontrada
```

```python
def traducir(palabra_ing, idioma):
    try:
        return traducciones[palabra_ing][idioma]
    except KeyError:
        return "Traducción no encontrada"

print("TRADUCTOR SIMPLE")

print("Hello en español:", traducir("hello", "es"))
print("Hello en francés:", traducir("hello", "fr"))
print("World en portugués:", traducir("world", "pt"))
print("Goodbye en francés:", traducir("goodbye", "fr"))
```

Por todo esto los diccionarios son ideales para:

* Mapear relaciones complejas: Como palabras y sus traducciones.
* Acceso rápido: Recuperar valores utilizando claves únicas.
* Flexibilidad: Las claves pueden ser de tipos inmutables, y los valores pueden ser de cualquier tipo, incluyendo otros diccionarios.

# Conjuntos
Problema: Eliminación de Elementos Duplicados en Grandes Volúmenes de Datos
Estás analizando grandes volúmenes de datos donde necesitas identificar elementos únicos y eliminar duplicados de manera eficiente. Por ejemplo, tienes una lista con miles de direcciones de correo electrónico de usuarios registrados en tu plataforma @1001problemas.com y necesitas obtener una lista de correos únicos para enviar una notificación. ¿Qué estructura de datos puedes usar para lograr esto de forma óptima?

### Limitaciones de las Listas
Las listas permiten elementos duplicados y eliminar duplicados manualmente es ineficiente para grandes volúmenes de datos.

### Definición y uso de conjuntos
Los conjuntos en Python son colecciones desordenadas de elementos únicos. Se definen utilizando llaves {} o la función set().

### Creación de un Conjunto

```python
# Lista de correos electrónicos con duplicados
correos = [
    "usuario1@1001problemas.com",
    "usuario2@1001problemas.com",
    "usuario3@1001problemas.com",
    "usuario1@1001problemas.com",  # Duplicado
    "usuario4@1001problemas.com",
    "usuario2@1001problemas.com",  # Duplicado
]

# Convertir la lista en un conjunto para eliminar duplicados
correos_unicos = set(correos)
print(correos_unicos)
```

```python
correos = [
    "usuario1@1001problemas.com",
    "usuario2@1001problemas.com",
    "usuario3@1001problemas.com",
    "usuario1@1001problemas.com",
    "usuario4@1001problemas.com",
    "usuario2@1001problemas.com",
    "usuario5@1001problemas.com"
]

correos_unicos = set(correos)

print("Lista original:")
print(correos)

print("\nConjunto sin duplicados:")
print(correos_unicos)
```

### Características de los Conjuntos
* Elementos Únicos: No permite duplicados.
* Desordenados: No mantienen un orden específico.
* Operaciones Matemáticas: Permite realizar operaciones como unión, intersección y diferencia.

### Operaciones 
* Unión

Combina elementos de ambos conjuntos sin duplicados.

```python
suscriptores_newsletter = {
    "usuario1@1001problemas.com",
    "usuario5@1001problemas.com",
    "usuario6@1001problemas.com",
}

todos_correos = correos_unicos.union(suscriptores_newsletter)
print(todos_correos)
```

```python
suscriptores_newsletter = {
    "usuario2@1001problemas.com",
    "usuario5@1001problemas.com",
    "usuario6@1001problemas.com",
    "usuario7@1001problemas.com"
}

todos_correos = correos_unicos.union(suscriptores_newsletter)

print("Correos registrados:")
print(correos_unicos)

print("\nSuscriptores del boletín:")
print(suscriptores_newsletter)

print("\nTodos los correos sin repetir:")
print(todos_correos)
```

* Intersección

Obtiene los elementos comunes entre conjuntos.

```python
correos_comunes = correos_unicos.intersection(suscriptores_newsletter)
print(correos_comunes)
```

```python
correos_comunes = correos_unicos.intersection(suscriptores_newsletter)

print("Correos que aparecen en ambos conjuntos:")
print(correos_comunes)

print("\nCantidad de correos comunes:", len(correos_comunes))
```

* Diferencia

Elementos que están en el primer conjunto pero no en el segundo.

```python
solo_registrados = correos_unicos.difference(suscriptores_newsletter)
print(solo_registrados)
```

```python
solo_registrados = correos_unicos.difference(suscriptores_newsletter)

print("Correos registrados que no están suscritos:")
print(solo_registrados)

print("Cantidad de correos:", len(solo_registrados))
```

### Métodos útiles de los Conjuntos
* add(): Agrega un elemento al conjunto.

```python
correos_unicos.add("usuario7@1001problemas.com")
print(correos_unicos)
```

```python
print("Correos antes de agregar:")
print(correos_unicos)

correos_unicos.add("usuario8@1001problemas.com")

print("\nCorreos después de agregar:")
print(correos_unicos)
```

* remove(): Elimina un elemento; lanza un error si no existe.

```python
correos_unicos.remove("usuario4@1001problemas.com")
print(correos_unicos)
```

```python
print("Conjunto antes de eliminar:")
print(correos_unicos)

correos_unicos.remove("usuario8@1001problemas.com")

print("\nConjunto después de eliminar:")
print(correos_unicos)
```

* discard(): Elimina un elemento si existe; no lanza error si no está.

```python
correos_unicos.discard("usuario10@1001problemas.com")
```

```python
print("Conjunto original:")
print(correos_unicos)

correos_unicos.discard("usuario10@1001problemas.com")

print("\nConjunto después de utilizar discard:")
print(correos_unicos)
```

* clear(): Elimina todos los elementos.

```python
correos_unicos.clear()
print(correos_unicos)  # Salida: set()
```

```python
print("Conjunto antes de vaciar:")
print(correos_unicos)

correos_unicos.clear()

print("\nConjunto después de vaciar:")
print(correos_unicos)

print("Cantidad de elementos:", len(correos_unicos))
```

```python
# DESAFÍO EXTRA: COORDENADAS DE SATÉLITES

satelites = {
    "Satélite 1": (3, 4, 5),
    "Satélite 2": (10, 8, 6),
    "Satélite 3": (2, 1, 3),
    "Satélite 4": (15, 12, 9),
    "Satélite 5": (7, 5, 4)
}

distancias = {}

print("COORDENADAS DE LOS SATÉLITES")

for nombre, coordenadas in satelites.items():
    x, y, z = coordenadas

    distancia = (x**2 + y**2 + z**2)**0.5
    distancias[nombre] = distancia

    print(nombre, "- Coordenadas:", coordenadas)
    print("Distancia al origen:", round(distancia, 2))

mas_cercano = min(distancias, key=distancias.get)
mas_lejano = max(distancias, key=distancias.get)

print("\nRESULTADOS")
print("Satélite más cercano:", mas_cercano)
print("Distancia:", round(distancias[mas_cercano], 2))

print("Satélite más lejano:", mas_lejano)
print("Distancia:", round(distancias[mas_lejano], 2))
```

```python

# DESAFÍO 93: TRADUCTOR MULTILINGÜE

traducciones = {
    "hello": {
        "es": "hola",
        "fr": "bonjour",
        "it": "ciao",
        "de": "hallo"
    },
    "world": {
        "es": "mundo",
        "fr": "monde",
        "it": "mondo",
        "de": "Welt"
    },
    "goodbye": {
        "es": "adiós",
        "fr": "au revoir",
        "it": "arrivederci",
        "de": "auf Wiedersehen"
    },
    "thanks": {
        "es": "gracias",
        "fr": "merci",
        "it": "grazie",
        "de": "danke"
    }
}

def traducir(palabra, idiomas):
    print("\nPalabra:", palabra)

    if palabra in traducciones:
        for idioma in idiomas:
            if idioma in traducciones[palabra]:
                print(idioma, ":", traducciones[palabra][idioma])
            else:
                print(idioma, ": Traducción no encontrada")
    else:
        print("La palabra no está en el diccionario.")

print("TRADUCTOR MULTILINGÜE")

traducir("hello", ["es", "fr", "it", "de"])
traducir("thanks", ["es", "it", "de"])
traducir("computer", ["es", "fr"])
```

```python

# DESAFÍO 94: GESTIÓN DE USUARIOS

astronomia = {
    "usuario1@1001problemas.com",
    "usuario2@1001problemas.com",
    "usuario3@1001problemas.com",
    "usuario4@1001problemas.com"
}

matematica = {
    "usuario2@1001problemas.com",
    "usuario3@1001problemas.com",
    "usuario5@1001problemas.com"
}

programacion = {
    "usuario2@1001problemas.com",
    "usuario3@1001problemas.com",
    "usuario6@1001problemas.com",
    "usuario7@1001problemas.com"
}

todos_los_concursos = astronomia.intersection(
    matematica, programacion
)

solo_programacion = programacion.difference(
    astronomia.union(matematica)
)

usuarios_unicos = astronomia.union(
    matematica, programacion
)

print("GESTIÓN DE CONCURSOS")

print("\nParticiparon en los tres concursos:")
print(sorted(todos_los_concursos))

print("\nParticiparon solamente en Programación:")
print(sorted(solo_programacion))

print("\nTodos los usuarios únicos:")
print(sorted(usuarios_unicos))

print("\nCantidad total de participantes:", len(usuarios_unicos))
```

```python

# DESAFÍO 95: ANÁLISIS DE VENTAS

ventas = [
    (101, 3, 250),
    (102, 5, 120),
    (103, 2, 500),
    (101, 4, 250),
    (102, 3, 120),
    (104, 6, 80)
]

ingreso_total = 0
cantidades = {}

print("REGISTRO DE VENTAS")

for venta in ventas:
    codigo, cantidad, precio = venta

    subtotal = cantidad * precio
    ingreso_total += subtotal

    print("Producto:", codigo)
    print("Cantidad:", cantidad)
    print("Subtotal: $", subtotal)

    if codigo in cantidades:
        cantidades[codigo] += cantidad
    else:
        cantidades[codigo] = cantidad

producto_mas_vendido = max(cantidades, key=cantidades.get)

print("\nRESULTADOS")

print("Ingreso total: $", ingreso_total)

print("Cantidad vendida por producto:")
for codigo, cantidad in cantidades.items():
    print("Producto", codigo, ":", cantidad, "unidades")

print("\nProducto más vendido:", producto_mas_vendido)
print("Unidades vendidas:", cantidades[producto_mas_vendido])
```

```python

# DESAFÍO 96: INVENTARIO DE PRODUCTOS

inventario = {
    101: {"nombre": "Cuaderno", "precio": 120, "stock": 15},
    102: {"nombre": "Lápiz", "precio": 25, "stock": 40},
    103: {"nombre": "Mochila", "precio": 850, "stock": 8}
}

def agregar_producto(codigo, nombre, precio, stock):
    if codigo not in inventario:
        inventario[codigo] = {
            "nombre": nombre,
            "precio": precio,
            "stock": stock
        }
        print("Producto agregado:", nombre)
    else:
        print("El producto ya existe.")

def actualizar_stock(codigo, cantidad):
    if codigo in inventario:
        inventario[codigo]["stock"] = cantidad
        print("Stock actualizado correctamente.")
    else:
        print("Producto no encontrado.")

def buscar_producto(codigo):
    if codigo in inventario:
        print("\nProducto encontrado:")
        print("Código:", codigo)
        print("Nombre:", inventario[codigo]["nombre"])
        print("Precio: $", inventario[codigo]["precio"])
        print("Stock:", inventario[codigo]["stock"])
    else:
        print("Producto no encontrado.")

print("SISTEMA DE INVENTARIO")

agregar_producto(104, "Cartuchera", 180, 12)

actualizar_stock(102, 35)

buscar_producto(102)
buscar_producto(104)
buscar_producto(999)

print("\nINVENTARIO COMPLETO")

for codigo, producto in inventario.items():
    print(codigo, "-", producto)
```

```python

# DESAFÍO 97: REDES SOCIALES

facebook = {"Ana", "Luis", "Pedro", "Sofía", "Martín"}
instagram = {"Ana", "Luis", "Lucía", "Sofía", "Carla"}
twitter = {"Ana", "Pedro", "Lucía", "Sofía", "Diego"}

tres_plataformas = facebook.intersection(instagram, twitter)

solo_facebook = facebook.difference(instagram.union(twitter))
solo_instagram = instagram.difference(facebook.union(twitter))
solo_twitter = twitter.difference(facebook.union(instagram))

solo_una = solo_facebook.union(solo_instagram, solo_twitter)

al_menos_dos = facebook.intersection(instagram).union(
    facebook.intersection(twitter),
    instagram.intersection(twitter)
)

print("ANÁLISIS DE REDES SOCIALES")

print("\nUsuarios de las tres plataformas:")
print(sorted(tres_plataformas))

print("\nUsuarios de una sola plataforma:")
print(sorted(solo_una))

print("\nUsuarios de al menos dos plataformas:")
print(sorted(al_menos_dos))
```

```python

# DESAFÍO 99: SIMULACIÓN CLIMÁTICA

clima = [
    ("Montevideo", 22, 65),
    ("Buenos Aires", 25, 70),
    ("Santiago", 18, 55),
    ("Madrid", 30, 40),
    ("Londres", 15, 80)
]

ciudad_calida = clima[0]
ciudad_fria = clima[0]

humedad_total = 0

print("REGISTROS CLIMÁTICOS")

for registro in clima:
    ciudad, temperatura, humedad = registro

    print("\nCiudad:", ciudad)
    print("Temperatura:", temperatura, "°C")
    print("Humedad:", humedad, "%")

    humedad_total += humedad

    if temperatura > ciudad_calida[1]:
        ciudad_calida = registro

    if temperatura < ciudad_fria[1]:
        ciudad_fria = registro

promedio_humedad = humedad_total / len(clima)

print("\nRESULTADOS")

print("Ciudad más cálida:", ciudad_calida[0])
print("Temperatura:", ciudad_calida[1], "°C")

print("Ciudad más fría:", ciudad_fria[0])
print("Temperatura:", ciudad_fria[1], "°C")

print("Promedio de humedad:", round(promedio_humedad, 2), "%")
```

```python

# DESAFÍO 100: ANÁLISIS DEPORTIVO

torneo_1 = {"Ana", "Luis", "Pedro", "Sofía", "Martín"}
torneo_2 = {"Ana", "Luis", "Lucía", "Sofía", "Carla"}
torneo_3 = {"Ana", "Pedro", "Sofía", "Diego", "Lucía"}

todos_torneos = torneo_1.intersection(torneo_2, torneo_3)

solo_torneo_1 = torneo_1.difference(torneo_2.union(torneo_3))
solo_torneo_2 = torneo_2.difference(torneo_1.union(torneo_3))
solo_torneo_3 = torneo_3.difference(torneo_1.union(torneo_2))

un_solo_torneo = solo_torneo_1.union(
    solo_torneo_2, solo_torneo_3
)

atletas_unicos = torneo_1.union(torneo_2, torneo_3)

print("ANÁLISIS DE TORNEOS DEPORTIVOS")

print("\nAtletas que participaron en todos los torneos:")
print(sorted(todos_torneos))

print("\nAtletas que participaron en un solo torneo:")
print(sorted(un_solo_torneo))

print("\nTodos los atletas participantes:")
print(sorted(atletas_unicos))

print("\nCantidad total de atletas únicos:", len(atletas_unicos))
```
