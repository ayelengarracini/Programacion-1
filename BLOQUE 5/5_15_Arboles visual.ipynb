# Árboles en Python

Los árboles son una estructura de datos fundamental en ciencias de la computación y programación. Un árbol consiste en un conjunto de nodos conectados por bordes, con un nodo raíz desde el cual se derivan los demás nodos. Cada nodo puede tener cero o más nodos hijos.

En esta guía, exploraremos cómo implementar y trabajar con árboles en Python.

## Definición de un Nodo

Primero, definiremos la clase `Nodo` que representará cada nodo en nuestro árbol.


```python
class Nodo:
    def __init__(self, key):
        self.izq = None
        self.der = None
        self.val = key
```

```python

class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

nodo1 = Nodo(15)

print("Valor del nodo:", nodo1.valor)
print("Hijo izquierdo:", nodo1.izq)
print("Hijo derecho:", nodo1.der)
```

## Creación de un Árbol

A continuación, crearemos un árbol simple para ilustrar cómo se pueden conectar los nodos.


```python
# Crear el nodo raíz
raiz = Nodo(1)

# Añadir nodos hijos
raiz.izq = Nodo(2)
raiz.der = Nodo(3)

# Añadir nodos hijos al nodo izquierdo del raíz
raiz.izq.izq = Nodo(4)
raiz.izq.der = Nodo(5)

# Añadir nodos hijos al nodo derecho del raíz
raiz.der.izq = Nodo(6)
raiz.der.der = Nodo(7)

# Añadir nodos hijos al nodo hijo derecho del hijo izquierdo del raíz
raiz.izq.der.izq = Nodo(8)
raiz.izq.der.der = Nodo(9)

# Añadir nodos hijos al nodo hijo derecho del hijo derecho del raíz
raiz.der.der.izq = Nodo(10)
raiz.der.der.der = Nodo(11)
```

En el codigo anterior se creo el árbol de ejemplo inicial, en principio cargandolo por _niveles_. Veremos mas adelante otras formas de recorrer un árbol.

```python

raiz = Nodo(10)

raiz.izq = Nodo(5)
raiz.der = Nodo(15)

raiz.izq.izq = Nodo(3)
raiz.izq.der = Nodo(7)

raiz.der.izq = Nodo(12)
raiz.der.der = Nodo(18)

print("Raíz del árbol:", raiz.valor)
print("Hijo izquierdo:", raiz.izq.valor)
print("Hijo derecho:", raiz.der.valor)

print("\nNodos del lado izquierdo:")
print(raiz.izq.izq.valor)
print(raiz.izq.der.valor)

print("\nNodos del lado derecho:")
print(raiz.der.izq.valor)
print(raiz.der.der.valor)
```

## Travesía de un Árbol

Existen diferentes maneras de recorrer un árbol. Aquí mostraremos tres métodos comunes: preorden, inorden y postorden.


### Travesía en Preorden

En la travesía en preorden, visitamos primero la raíz, luego el subárbol izquierdo, y finalmente el subárbol derecho.


```python
def print_preorder(raiz):
    if raiz:
        print(raiz.val, end=" ")
        print_preorder(raiz.izq)
        print_preorder(raiz.der)
        
print_preorder(raiz)  # Salida esperada: 1 2 4 5 8 9 3 6 7 10 11 
```

```python

def preorden(nodo):
    if nodo:
        print(nodo.valor, end=" ")
        preorden(nodo.izq)
        preorden(nodo.der)

print("Recorrido en preorden:")
preorden(raiz)

print("\nRaíz del árbol:", raiz.valor)
```

### Travesía en Inorden

En la travesía en inorden, visitamos primero el subárbol izquierdo, luego la raíz, y finalmente el subárbol derecho.


```python
def print_inorder(raiz):
    if raiz:
        print_inorder(raiz.izq)
        print(raiz.val, end=" ")
        print_inorder(raiz.der)
        
print_inorder(raiz)  # Salida esperada: 4 2 8 5 9 1 6 3 10 7 11
```

```python

def inorden(nodo):
    if nodo:
        inorden(nodo.izq)
        print(nodo.valor, end=" ")
        inorden(nodo.der)

print("Recorrido en inorden:")
inorden(raiz)

print("\nValor de la raíz:", raiz.valor)
```

### Travesía en Postorden

En la travesía en postorden, visitamos primero el subárbol izquierdo, luego el subárbol derecho, y finalmente la raíz.


```python
def print_postorder(raiz):
    if raiz:
        print_postorder(raiz.izq)
        print_postorder(raiz.der)
        print(raiz.val, end=" ")
        
print_postorder(raiz)  # Salida esperada: 4 8 9 5 2 6 10 11 7 3 1 
```

```python

def postorden(nodo):
    if nodo:
        postorden(nodo.izq)
        postorden(nodo.der)
        print(nodo.valor, end=" ")

print("Recorrido en postorden:")
postorden(raiz)

print("\nÚltimo nodo del recorrido:", raiz.valor)
```

```python
#DESAFÍO 78

class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

raiz = Nodo(40)
raiz.izq = Nodo(25)
raiz.der = Nodo(60)

raiz.izq.izq = Nodo(10)
raiz.izq.der = Nodo(85)
raiz.der.izq = Nodo(50)
raiz.der.der = Nodo(70)

def maximo_postorden(nodo):
    if nodo is None:
        return None

    max_izq = maximo_postorden(nodo.izq)
    max_der = maximo_postorden(nodo.der)

    print(nodo.valor, end=" ")

    mayor = nodo.valor

    if max_izq is not None and max_izq > mayor:
        mayor = max_izq

    if max_der is not None and max_der > mayor:
        mayor = max_der

    return mayor

print("RECORRIDO EN POSTORDEN")

mayor = maximo_postorden(raiz)

print("\n\nEl número mayor del árbol es:", mayor)
```

```python
#DESAFÍO 79

class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

def insertar(raiz, valor):
    if raiz is None:
        return Nodo(valor)

    if valor < raiz.valor:
        raiz.izq = insertar(raiz.izq, valor)
    elif valor > raiz.valor:
        raiz.der = insertar(raiz.der, valor)

    return raiz

def buscar(raiz, numero):
    if raiz is None:
        return False

    if raiz.valor == numero:
        return True

    if numero < raiz.valor:
        return buscar(raiz.izq, numero)
    else:
        return buscar(raiz.der, numero)

numeros = [50, 30, 70, 20, 40, 60, 80]

raiz = None

for numero in numeros:
    raiz = insertar(raiz, numero)

print("ÁRBOL DE BÚSQUEDA BINARIA")
print("Números insertados:", numeros)

busquedas = [40, 90, 20, 65]

for numero in busquedas:
    if buscar(raiz, numero):
        print(f"El número {numero} está en el árbol.")
    else:
        print(f"El número {numero} no se encuentra.")
```

```python
#DESAFÍO 80

import ast
import operator

class Nodo:
    def __init__(self, valor, izq=None, der=None):
        self.valor = valor
        self.izq = izq
        self.der = der

operaciones = {
    ast.Add: operator.add,
    ast.Sub: operator.sub,
    ast.Mult: operator.mul,
    ast.Div: operator.truediv
}

def construir_arbol(expresion):
    expresion = ast.parse(expresion, mode="eval")
    return convertir_nodo(expresion.body)

def convertir_nodo(nodo):
    if isinstance(nodo, ast.Constant) and type(nodo.value) is int:
        return Nodo(nodo.value)

    if isinstance(nodo, ast.UnaryOp) and isinstance(nodo.op, ast.USub):
        return Nodo("-", Nodo(0), convertir_nodo(nodo.operand))

    if isinstance(nodo, ast.BinOp) and type(nodo.op) in operaciones:
        simbolos = {
            ast.Add: "+",
            ast.Sub: "-",
            ast.Mult: "*",
            ast.Div: "/"
        }

        simbolo = simbolos[type(nodo.op)]

        return Nodo(
            simbolo,
            convertir_nodo(nodo.left),
            convertir_nodo(nodo.right)
        )

    raise ValueError("La expresión contiene elementos no permitidos.")

def evaluar(nodo):
    if nodo.izq is None and nodo.der is None:
        return nodo.valor

    izquierda = evaluar(nodo.izq)
    derecha = evaluar(nodo.der)

    simbolos = {
        "+": operator.add,
        "-": operator.sub,
        "*": operator.mul,
        "/": operator.truediv
    }

    return simbolos[nodo.valor](izquierda, derecha)

expresion = "5 + 3 * 4"

arbol = construir_arbol(expresion)
resultado = evaluar(arbol)

print("EVALUACIÓN DE EXPRESIONES")
print("Expresión:", expresion)
print("Resultado:", resultado)

print("\nOtra prueba:")

otra_expresion = "(10 + 5) * 2"
otro_arbol = construir_arbol(otra_expresion)

print("Expresión:", otra_expresion)
print("Resultado:", evaluar(otro_arbol))
```
