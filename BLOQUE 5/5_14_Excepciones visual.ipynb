# Excepciones y errores

## Excepciones

"El agua hierve a 100 grados bajo cero cuando hace calor."

Aunque esta oración es correcta desde el punto de vista gramatical, su significado está lleno de contradicciones. Algo similar ocurre en la programación: incluso cuando el código tiene la sintaxis correcta, puede generar errores lógicos durante la ejecución. A estos errores se les conoce como excepciones, y son detectados cuando el programa se está ejecutando, generando mensajes de error que pueden, en muchos casos, ser controlados y gestionados.

Ejecuta las siguientes líneas y observa que ocurre...

```python
10 * (1/0)
```

```python
4 + spam*3
```

```python
'2' + 2
```

La última línea del mensaje de error es crucial, ya que indica qué tipo de excepción ha ocurrido. Cada tipo de excepción tiene un propósito específico y se utiliza para señalar un tipo particular de error.

Explicar las caracteristicas propias de los errores cometidos en los codigos anteriores.

```python

print("===== EXCEPCIONES Y ERRORES =====")

try:
    resultado = 15 / 0
    print(resultado)
except ZeroDivisionError:
    print("Error: no se puede dividir entre cero.")

print()

try:
    resultado = 5 + numero
    print(resultado)
except NameError:
    print("Error: la variable no está definida.")

print()

try:
    resultado = "Hola" + 10
    print(resultado)
except TypeError:
    print("Error: no se puede sumar un texto con un número.")

print()

print("===== OPERACIONES CORRECTAS =====")

print("División:", 15 / 3)

numero = 4
print("Suma:", 5 + numero)

print("Texto:", "Hola " + str(10))
```

```python
#DESAFÍO 67
pruebas = [(10, 2), (10, 0), ("hola", 5)]

for numero1, numero2 in pruebas:
    try:
        a = int(numero1)
        b = int(numero2)

        resultado = a / b
        print("El resultado es:", resultado)

    except ZeroDivisionError:
        print("Error: No se puede dividir entre cero.")

    except ValueError:
        print("Error: Debes ingresar números enteros.")

    print()
```

```python
#DESAFÍO 68

valores = [10, 5, 0, "hola", "8", None]

print("OPERACIONES MATEMÁTICAS")

for valor in valores:
    try:
        numero = int(valor)

        resultado = 100 / numero
        print("100 dividido entre", numero, "=", resultado)

    except ZeroDivisionError:
        print("Error: División entre cero.")

    except ValueError:
        print("Error: El valor no es numérico.")

    except TypeError:
        print("Error: Tipo de dato incorrecto.")
```

```python
#DESAFÍO 69

def factorial(numero):
    if type(numero) != int:
        raise TypeError("El número debe ser entero.")

    if numero < 0:
        raise ValueError("El número no puede ser negativo.")

    if numero > 100:
        raise OverflowError("El número es demasiado grande para este ejercicio.")

    resultado = 1

    for i in range(1, numero + 1):
        resultado = resultado * i

    return resultado


numeros = [5, 4, -3, 2.5, 150]

for numero in numeros:
    try:
        print("Factorial de", numero, "=", factorial(numero))

    except (TypeError, ValueError, OverflowError) as error:
        print("Error:", error)
```

```python
#DESAFÍO 70

import math


class NegativeNumberError(Exception):
    pass


numeros = [25, 16, -9, "hola"]

for numero in numeros:
    try:
        valor = float(numero)

        if valor < 0:
            raise NegativeNumberError("No se puede calcular la raíz de un número negativo.")

        resultado = math.sqrt(valor)
        print("Raíz cuadrada de", valor, "=", resultado)

    except NegativeNumberError as error:
        print("Error:", error)

    except ValueError:
        print("Error: Debes ingresar un número.")

    print()
```
