# ¿Qué es un arreglo unidimensional?

En términos muy sencillos, un arreglo es una colección de elementos que son *del mismo tipo de datos*. Piensa en él como una estantería donde cada estante tiene un valor que tú decides.

```python
productos = ["manzanas", "bananas", "cerezas", "duraznos", "peras"]
```

Aquí, productos es un arreglo (en Python, lo llamamos "lista") que contiene varias cadenas.

*Nota:*
*Los clásicos arrays descritos en lenguajes como C, Java, entre otros no existen en Python como tal. Una forma elegante de superar esta limitación es utilizar las llamadas listas in Python, que tienen una función similar. A diferencia de los contenedores de Java, las listas de Python pueden contener diferentes tipos de valores.*

# Eliminar elementos del array con pop ( ) o remove ( )

# Lista de diferentes métodos para arrays en Python

Python ofrece muchos métodos útiles para trabajar con listas. Aquí hay una tabla con algunos de los más comunes:

|Método|Descripción|
|---|---|
|`append()`|Añade un elemento al final de la lista.|
|`clear()`|Este método elimina todos los elementos de la lista.|
|`copy()`|`copy()` produce una copia de toda la lista.|
|`count()`|Este método da como resultado el número exacto de elementos con un valor determinado.|
|`extend()`|`extend()` añade todos los elementos de una lista al final de una lista en Python.|
|`index()`|Muestra el número de índice del primer elemento con un valor determinado.|
|`insert()`|Añade un elemento en una posición determinada.|
|`len()`|Con `len()` se determina la longitud de un array en Python.|
|`pop()`|Con `pop()` se elimina un elemento en una posición determinada.|
|`remove()`|`remove()` elimina el primer elemento con un valor determinado.|
|`reverse()`|Utiliza este método para invertir el orden de los elementos de tu lista en Python.|
|`sort()`|Con `sort()` puedes ordenar tu lista.|


```python

# =====================================
# OPERACIONES CON LISTAS
# =====================================

numeros = [3, 5, 8, 12, 20, 7, 15, 4]

print("===== ANÁLISIS DE NÚMEROS =====")

print("Lista original:", numeros)

# Operaciones matemáticas
print("Suma total:", sum(numeros))
print("Número mayor:", max(numeros))
print("Número menor:", min(numeros))

promedio = sum(numeros) / len(numeros)
print(f"Promedio: {promedio:.2f}")

# Mostrar los números pares
print("\nNúmeros pares:")

for numero in numeros:
    if numero % 2 == 0:
        print(numero)

# Calcular diferencias consecutivas
print("\nDiferencias entre números:")

for i in range(len(numeros) - 1):
    diferencia = numeros[i + 1] - numeros[i]
    print(f"{numeros[i + 1]} - {numeros[i]} = {diferencia}")

# Ordenar los números
print("\nLista ordenada:")
print(sorted(numeros))
```

```python

# =====================================
# POSICIONES DE UNA LISTA
# =====================================

mi_lista = [1, 2, 3]

print("===== OPERACIONES CON POSICIONES =====")

print("Lista original:", mi_lista)

# Acceder a los elementos
print("Primera posición:", mi_lista[0])
print("Segunda posición:", mi_lista[1])
print("Tercera posición:", mi_lista[2])

# Realizar operaciones
suma = mi_lista[0] + mi_lista[1] + mi_lista[2]
multiplicacion = mi_lista[0] * mi_lista[1] * mi_lista[2]

print("\nSuma de los elementos:", suma)
print("Multiplicación de los elementos:", multiplicacion)

# Modificar un elemento
mi_lista[1] = 10

print("\nLista modificada:", mi_lista)
print("Nueva suma:", sum(mi_lista))
```

```python

# ==============================================
# DESAFÍO 18: GESTIÓN DEL STOCK DE LA VERDULERÍA
# ==============================================

print("===== VERDULERÍA DE AYELÉN =====")
print("Bienvenida al sistema de inventario\n")

# INVENTARIO INICIAL
inventario = [
    "manzanas", "bananas", "zanahorias",
    "espinacas", "brocoli", "cebolla", "kiwis"
]

print("INVENTARIO INICIAL:")
print(inventario)

# ==============================================
# PREGUNTA 1: ¿CUÁNTOS PRODUCTOS HAY?
# ==============================================

cantidad = len(inventario)

print("\nPREGUNTA 1:")
print(f"Hay {cantidad} tipos de productos.")

# ==============================================
# PREGUNTA 2: PRODUCTO EN TERCERA POSICIÓN
# ==============================================

print("\nPREGUNTA 2:")
print("El tercer producto es:", inventario[2])

# ==============================================
# PREGUNTA 3: ELIMINAR LAS BANANAS
# ==============================================

inventario.remove("bananas")

print("\nPREGUNTA 3:")
print("Se vendieron todas las bananas.")
print("Inventario actualizado:", inventario)

# ==============================================
# PREGUNTA 4: AGREGAR NUEVOS PRODUCTOS
# ==============================================

nuevos_productos = ["frutillas", "apio", "papas"]

inventario.extend(nuevos_productos)

print("\nPREGUNTA 4:")
print("Llegaron nuevos productos.")
print("Inventario actualizado:", inventario)

# ==============================================
# PREGUNTA 5: VERIFICAR SI HAY PAPAS
# ==============================================

print("\nPREGUNTA 5:")

if "papas" in inventario:
    print("Sí, tenemos papas disponibles.")
else:
    print("No tenemos papas disponibles.")

# ==============================================
# PREGUNTA 6: HACER ESPACIO PARA DRAGONFRUIT
# ==============================================

print("\nPREGUNTA 6:")

# La tienda solamente admite 7 productos
capacidad_maxima = 7

# Elegimos qué productos retirar
productos_a_retirar = ["espinacas", "brocoli"]

for producto in productos_a_retirar:
    if producto in inventario:
        inventario.remove(producto)
        print(f"Se retiró: {producto}")

# Si todavía no hay espacio, retiramos
# productos del final del inventario
while len(inventario) >= capacidad_maxima:
    producto_retirado = inventario.pop()
    print(f"Se retiró para hacer espacio: {producto_retirado}")

# Agregamos la fruta nueva
inventario.append("dragonfruit")

print("Se agregó dragonfruit.")
print("Inventario actualizado:", inventario)
print("Cantidad de productos:", len(inventario))

# ==============================================
# PREGUNTA 7: ORDENAR ALFABÉTICAMENTE
# ==============================================

inventario.sort()

print("\nPREGUNTA 7:")
print("Inventario ordenado alfabéticamente:")

for posicion, producto in enumerate(inventario, start=1):
    print(f"{posicion}. {producto}")

# ==============================================
# PREGUNTA 8: CREAR UNA COPIA INDEPENDIENTE
# ==============================================

copia_inventario = inventario.copy()

print("\nPREGUNTA 8:")
print("Copia entregada al nuevo empleado:")
print(copia_inventario)

# El empleado modifica su copia
copia_inventario.append("sandías")

print("\nEl empleado agregó sandías a su copia.")

print("Inventario original:")
print(inventario)

print("Copia modificada:")
print(copia_inventario)

# ==============================================
# RESUMEN FINAL
# ==============================================

print("\n===== RESUMEN FINAL =====")
print("Verdulería de Ayelén")

print(f"Productos disponibles: {len(inventario)}")
print(f"Capacidad máxima: {capacidad_maxima}")

if len(inventario) <= capacidad_maxima:
    print("El inventario respeta la capacidad.")
else:
    print("Atención: se superó la capacidad.")

print("\n¡Inventario gestionado correctamente!")
```

```python

# ==========================================
# DESAFÍO 20: CARACTERES ASCII
# ==========================================

print("===== CONVERTIDOR ASCII =====")

# Lista de códigos ASCII
codigos = [72, 79, 76, 65, 32, 65, 89, 69]

print("\nCódigos originales:")
print(codigos)

# Convertir los números a caracteres
print("\n===== CONVERSIÓN =====")

caracteres = []

for codigo in codigos:
    caracter = chr(codigo)
    caracteres.append(caracter)

    print(f"Código {codigo} = '{caracter}'")

# Unir los caracteres para formar un mensaje
mensaje = "".join(caracteres)

print("\nMensaje formado:")
print(mensaje)

# ==========================================
# EXPERIMENTO: TEXTO A CÓDIGOS ASCII
# ==========================================

print("\n===== EXPERIMENTO EXTRA =====")

nombre = "AYELEN"

print("Nombre:", nombre)
print("\nCódigos de cada letra:")

for letra in nombre:
    codigo = ord(letra)
    print(f"{letra} = {codigo}")

# Crear una lista con los códigos
codigos_nombre = [ord(letra) for letra in nombre]

print("\nLista de códigos del nombre:")
print(codigos_nombre)

# Reconstruir el nombre
nombre_reconstruido = ""

for codigo in codigos_nombre:
    nombre_reconstruido += chr(codigo)

print("\nNombre reconstruido:")
print(nombre_reconstruido)

print("\n===== FIN DEL PROGRAMA =====")
```

```python

# ==========================================
# DESAFÍO 21: CONTROL DE NOTAS
# ==========================================

from collections import Counter
import matplotlib.pyplot as plt

print("===== CONTROL DE NOTAS =====")

# Lista de estudiantes
estudiantes = [
    "Ana", "Carlos", "Lucía", "Pedro",
    "María", "Sofía", "Juan", "Valentina"
]

# Notas de los estudiantes
notas = [10, 8, 12, 5, 9, 8, 6, 11]

# ==========================================
# MOSTRAR LAS CALIFICACIONES
# ==========================================

print("\n===== LISTA DE ESTUDIANTES =====")

for estudiante, nota in zip(estudiantes, notas):
    print(f"{estudiante}: {nota}")

# ==========================================
# PREGUNTA 1: CALCULAR EL PROMEDIO
# ==========================================

promedio = sum(notas) / len(notas)

print("\nPREGUNTA 1:")
print(f"Promedio de la clase: {promedio:.2f}")

# ==========================================
# PREGUNTA 2: NOTA MÁS ALTA Y MÁS BAJA
# ==========================================

nota_maxima = max(notas)
nota_minima = min(notas)

print("\nPREGUNTA 2:")
print("Nota más alta:", nota_maxima)
print("Nota más baja:", nota_minima)

# ==========================================
# PREGUNTA 3: NOTA MÁS FRECUENTE
# ==========================================

contador = Counter(notas)

frecuencia_maxima = max(contador.values())

notas_frecuentes = [
    nota for nota, frecuencia in contador.items()
    if frecuencia == frecuencia_maxima
]

print("\nPREGUNTA 3:")
print("Nota o notas más frecuentes:", notas_frecuentes)
print("Cantidad de repeticiones:", frecuencia_maxima)

# ==========================================
# EXPERIMENTO: APROBADOS Y DESAPROBADOS
# ==========================================

aprobados = 0
desaprobados = 0

print("\n===== RESULTADOS ACADÉMICOS =====")

for estudiante, nota in zip(estudiantes, notas):

    if nota >= 7:
        print(f"{estudiante}: APROBADO")
        aprobados += 1

    else:
        print(f"{estudiante}: DESAPROBADO")
        desaprobados += 1

print("\n===== RESUMEN =====")
print("Cantidad de estudiantes:", len(estudiantes))
print("Aprobados:", aprobados)
print("Desaprobados:", desaprobados)

# ==========================================
# PREGUNTA 4: GRÁFICO DE BARRAS
# ==========================================

plt.figure(figsize=(10, 6))

# Colores según aprobación
colores = [
    "mediumseagreen" if nota >= 7 else "salmon"
    for nota in notas
]

plt.bar(estudiantes, notas, color=colores)

plt.title("Calificaciones de los Estudiantes")
plt.xlabel("Estudiantes")
plt.ylabel("Calificaciones")

plt.ylim(0, 13)
plt.yticks(range(0, 13))

# Línea que indica la nota mínima aprobatoria
plt.axhline(
    y=7,
    color="red",
    linestyle="--",
    label="Mínimo para aprobar"
)

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

# Mostrar el gráfico
plt.show()

print("\n===== FIN DEL ANÁLISIS =====")
```

# Problema 1: Análisis de Ventas Mensuales usando el módulo array
Eres el gerente de una tienda de ropa y necesitas analizar las ventas mensuales de un producto específico durante el último año. Usarás el módulo array para almacenar y manipular los datos de ventas de cada mes. El objetivo es calcular la venta total anual, el promedio de ventas mensuales, y encontrar los meses con la venta más baja y más alta.

## Datos Iniciales
Las ventas mensuales del producto específico son las siguientes (en unidades):

```python
import array

ventas_mensuales = array.array('i', [120, 135, 150, 145, 160, 155, 130, 140, 125, 170, 160, 180])
```

## Tareas:
**Calcula la venta total anual.**

```python
total_anual = sum(ventas_mensuales)
print(f"Venta total anual: {total_anual} unidades")
```

**Calcula el promedio de ventas mensuales.**

```python
promedio_mensual = total_anual / len(ventas_mensuales)
print(f"Promedio de ventas mensuales: {promedio_mensual:.2f} unidades")
```

**Encuentra el mes con la venta más baja y más alta.**

```python
mes_baja = ventas_mensuales.index(min(ventas_mensuales)) + 1
mes_alta = ventas_mensuales.index(max(ventas_mensuales)) + 1
print(f"Mes con la venta más baja: {mes_baja} (Venta: {min(ventas_mensuales)} unidades)")
print(f"Mes con la venta más alta: {mes_alta} (Venta: {max(ventas_mensuales)} unidades)")
```

```python

# ==========================================
# PROBLEMA 1: ANÁLISIS DE VENTAS MENSUALES
# ==========================================

import array

print("===== TIENDA DE ROPA DE AYELÉN =====")
print("===== ANÁLISIS DE VENTAS ANUALES =====")

# Nombres de los meses
meses = [
    "Enero", "Febrero", "Marzo", "Abril",
    "Mayo", "Junio", "Julio", "Agosto",
    "Septiembre", "Octubre", "Noviembre", "Diciembre"
]

# Ventas mensuales del producto (en unidades)
ventas_mensuales = array.array(
    'i', [120, 135, 150, 145, 160, 155,
          130, 140, 125, 170, 160, 180]
)

# ==========================================
# MOSTRAR LAS VENTAS DE CADA MES
# ==========================================

print("\n===== VENTAS POR MES =====")

for mes, venta in zip(meses, ventas_mensuales):
    print(f"{mes}: {venta} unidades")

# ==========================================
# TAREA 1: CALCULAR VENTA TOTAL ANUAL
# ==========================================

total_anual = sum(ventas_mensuales)

print("\n===== VENTA TOTAL ANUAL =====")
print(f"Total vendido: {total_anual} unidades")

# ==========================================
# TAREA 2: CALCULAR PROMEDIO MENSUAL
# ==========================================

promedio_mensual = total_anual / len(ventas_mensuales)

print("\n===== PROMEDIO MENSUAL =====")
print(f"Promedio: {promedio_mensual:.2f} unidades")

# ==========================================
# TAREA 3: MES CON MENOR Y MAYOR VENTA
# ==========================================

venta_minima = min(ventas_mensuales)
venta_maxima = max(ventas_mensuales)

# Buscamos las posiciones utilizando enumerate()
meses_bajos = [
    meses[i] for i, venta in enumerate(ventas_mensuales)
    if venta == venta_minima
]

meses_altos = [
    meses[i] for i, venta in enumerate(ventas_mensuales)
    if venta == venta_maxima
]

print("\n===== VENTAS MÍNIMAS Y MÁXIMAS =====")

print(f"Mes con menos ventas: {', '.join(meses_bajos)}")
print(f"Cantidad vendida: {venta_minima} unidades")

print(f"\nMes con más ventas: {', '.join(meses_altos)}")
print(f"Cantidad vendida: {venta_maxima} unidades")

# ==========================================
# EXPERIMENTO 1: MESES SOBRE EL PROMEDIO
# ==========================================

print("\n===== MESES QUE SUPERARON EL PROMEDIO =====")

for mes, venta in zip(meses, ventas_mensuales):
    if venta > promedio_mensual:
        print(f"{mes}: {venta} unidades")

# ==========================================
# EXPERIMENTO 2: CLASIFICAR LAS VENTAS
# ==========================================

print("\n===== CLASIFICACIÓN DE VENTAS =====")

for mes, venta in zip(meses, ventas_mensuales):

    if venta >= 160:
        estado = "VENTAS ALTAS"

    elif venta >= 130:
        estado = "VENTAS MEDIAS"

    else:
        estado = "VENTAS BAJAS"

    print(f"{mes}: {venta} unidades - {estado}")

# ==========================================
# EXPERIMENTO 3: SIMULAR UN AUMENTO
# ==========================================

print("\n===== SIMULACIÓN DE VENTAS =====")

# Simulamos un aumento del 10% en las ventas
ventas_simuladas = [
    round(venta * 1.10)
    for venta in ventas_mensuales
]

nuevo_total = sum(ventas_simuladas)

print(f"Total original: {total_anual} unidades")
print(f"Total simulado: {nuevo_total} unidades")
print(f"Diferencia: {nuevo_total - total_anual} unidades")

print("\n===== FIN DEL ANÁLISIS =====")
```

```python

# ==============================================
# PROBLEMA 2: ANÁLISIS DE TEMPERATURAS DIARIAS
# ==============================================

import numpy as np
import matplotlib.pyplot as plt

print("===== ANÁLISIS DE TEMPERATURAS =====")
print("===== REGISTRO DE 30 DÍAS =====")

# Temperaturas diarias registradas
temperaturas_diarias = np.array([
    22.5, 21.3, 23.1, 25.0, 24.5,
    22.1, 23.7, 24.8, 25.3, 26.1,
    23.4, 24.0, 22.9, 21.5, 23.0,
    24.3, 25.0, 24.7, 23.1, 22.4,
    23.5, 24.1, 25.4, 26.0, 24.8,
    23.9, 22.7, 23.3, 24.5, 25.1
])

dias = np.arange(1, 31)

# ==============================================
# TAREA 1: TEMPERATURA MEDIA
# ==============================================

temperatura_media = np.mean(temperaturas_diarias)

print("\n===== TEMPERATURA MEDIA =====")
print(f"Temperatura media: {temperatura_media:.2f} °C")

# ==============================================
# TAREA 2: TEMPERATURA MÁXIMA Y MÍNIMA
# ==============================================

temperatura_maxima = np.max(temperaturas_diarias)
temperatura_minima = np.min(temperaturas_diarias)

dia_maximo = np.argmax(temperaturas_diarias) + 1
dia_minimo = np.argmin(temperaturas_diarias) + 1

print("\n===== TEMPERATURAS EXTREMAS =====")

print(f"Temperatura máxima: {temperatura_maxima:.1f} °C")
print(f"Registrada el día: {dia_maximo}")

print(f"Temperatura mínima: {temperatura_minima:.1f} °C")
print(f"Registrada el día: {dia_minimo}")

# ==============================================
# TAREA 3: MOSTRAR LAS TEMPERATURAS
# ==============================================

print("\n===== TEMPERATURAS POR DÍA =====")

for dia, temperatura in zip(dias, temperaturas_diarias):
    print(f"Día {dia}: {temperatura:.1f} °C")

# ==============================================
# EXPERIMENTO 1: DÍAS SOBRE EL PROMEDIO
# ==============================================

dias_calidos = np.sum(
    temperaturas_diarias > temperatura_media
)

print("\n===== DÍAS SOBRE EL PROMEDIO =====")
print(f"Días por encima de la media: {dias_calidos}")

# ==============================================
# EXPERIMENTO 2: CLASIFICAR TEMPERATURAS
# ==============================================

print("\n===== CLASIFICACIÓN DEL CLIMA =====")

for dia, temperatura in zip(dias, temperaturas_diarias):

    if temperatura >= 25:
        estado = "Día cálido"

    elif temperatura >= 23:
        estado = "Día templado"

    else:
        estado = "Día fresco"

    print(f"Día {dia}: {temperatura:.1f} °C - {estado}")

# ==============================================
# EXPERIMENTO 3: VARIACIÓN DE TEMPERATURAS
# ==============================================

variaciones = np.diff(temperaturas_diarias)

print("\n===== CAMBIOS ENTRE DÍAS =====")

for i, cambio in enumerate(variaciones, start=2):

    if cambio > 0:
        print(f"Día {i}: Subió {cambio:.1f} °C")

    elif cambio < 0:
        print(f"Día {i}: Bajó {abs(cambio):.1f} °C")

    else:
        print(f"Día {i}: Sin cambios")

# ==============================================
# TAREA 4: GRÁFICO DE LÍNEAS
# ==============================================

plt.figure(figsize=(12, 6))

# Línea de temperaturas
plt.plot(
    dias,
    temperaturas_diarias,
    marker="o",
    color="royalblue",
    label="Temperatura diaria"
)

# Línea del promedio
plt.axhline(
    temperatura_media,
    color="orange",
    linestyle="--",
    label="Temperatura media"
)

# Marcar temperatura máxima
plt.scatter(
    dia_maximo,
    temperatura_maxima,
    color="red",
    s=100,
    label="Temperatura máxima"
)

# Marcar temperatura mínima
plt.scatter(
    dia_minimo,
    temperatura_minima,
    color="green",
    s=100,
    label="Temperatura mínima"
)

plt.title("Temperaturas Diarias del Último Mes")
plt.xlabel("Días del mes")
plt.ylabel("Temperatura (°C)")

plt.xticks(dias)
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()

print("\n===== FIN DEL ANÁLISIS =====")
```
