## Función Recursiva para Calcular la Potencia de un Número

En este ejercicio, vamos a crear una función recursiva que nos permita calcular la potencia de un número. La idea es que si tenemos un número \(a\) y queremos elevarlo a la potencia \(b\), la operación se representa como:

$$a^b = a \times a \times \dots \times a$$

Donde \(a\) se multiplica por sí mismo \(b\) veces.

Para implementar esto de manera recursiva, consideraremos algunos casos base:

1. Si \(b = 0\), cualquier número elevado a la potencia de 0 es 1.
2. Si la base \(a = 0\), el resultado es 0 sin importar el exponente.
3. Si \(b = 1\), el resultado es simplemente \(a\).

Con estos casos base en mente, podemos proceder a implementar la función.


```python
def potencia(a, b):
    # Caso base: cualquier número elevado a 0 es 1
    if b == 0:
        return 1
    # Caso base: 0 elevado a cualquier número es 0
    elif a == 0:
        return 0
    # Caso base: cualquier número elevado a 1 es él mismo
    elif b == 1:
        return a
    # Caso recursivo
    else:
        return a * potencia(a, b-1)

# Pruebas
print(potencia(2, 4))  # 16
print(potencia(4, 3))  # 64
print(potencia(3, 4))  # 81
```

```python

# FUNCIÓN RECURSIVA PARA CALCULAR POTENCIAS

def potencia(a, b):
    # Caso base: exponente igual a cero
    if b == 0:
        return 1

    # Caso base: exponente igual a uno
    elif b == 1:
        return a

    # Caso recursivo
    else:
        return a * potencia(a, b - 1)


# Pruebas del ejemplo original
print("===== POTENCIAS =====")

print("2 elevado a 4 =", potencia(2, 4))
print("4 elevado a 3 =", potencia(4, 3))
print("3 elevado a 4 =", potencia(3, 4))


# Experimentamos con otros números
print("\n===== OTROS EJEMPLOS =====")

print("5 elevado a 2 =", potencia(5, 2))
print("10 elevado a 3 =", potencia(10, 3))
print("7 elevado a 0 =", potencia(7, 0))


# Comparación con el operador de potencia de Python
print("\n===== COMPARACIÓN =====")

print("Con recursividad:", potencia(2, 5))
print("Con el operador **:", 2 ** 5)

print("\n===== FIN DEL PROGRAMA =====")
```

### Procesamiento de Texto Recursivo

Imaginemos que tenemos un texto con varias palabras, y queremos contar cuántas veces aparece una palabra específica de manera recursiva. Vamos a implementar una función recursiva que realice esta tarea. Para hacerlo más interesante, supongamos que estamos analizando una receta de cocina y queremos contar cuántas veces aparece la palabra "ingrediente" en la lista de instrucciones.

La idea es que vamos a analizar la receta como una lista de palabras, y recorreremos recursivamente la lista para contar cuántas veces aparece nuestra palabra objetivo.


```python
def contar_palabra(texto, palabra, indice=0):
    palabras = texto.split()  # Divide el texto en palabras
    if indice >= len(palabras):
        return 0  # Caso base: hemos terminado de revisar el texto
    # Caso recursivo: suma 1 si la palabra coincide, de lo contrario sigue buscando
    return (1 if palabras[indice] == palabra else 0) + contar_palabra(texto, palabra, indice + 1)

# Prueba de la función
texto_receta = "ingrediente harina ingrediente agua ingrediente sal ingrediente levadura"
print(contar_palabra(texto_receta, "ingrediente"))  # Debería devolver 4
```

```python

# PROCESAMIENTO DE TEXTO RECURSIVO

def contar_palabra(texto, palabra, indice=0):
    palabras = texto.lower().split()

    # Caso base: terminamos de revisar las palabras
    if indice >= len(palabras):
        return 0

    # Caso recursivo: contamos y seguimos
    if palabras[indice] == palabra.lower():
        return 1 + contar_palabra(texto, palabra, indice + 1)
    else:
        return contar_palabra(texto, palabra, indice + 1)


# Ejemplo original: receta de cocina
texto_receta = "ingrediente harina ingrediente agua ingrediente sal ingrediente levadura"

print("===== RECETA DE COCINA =====")
print("La palabra ingrediente aparece:",
      contar_palabra(texto_receta, "ingrediente"), "veces")


# Experimento 1: contar otra palabra
print("\n===== OTRO EJEMPLO =====")

texto = "Python es divertido y aprender Python es interesante"

print("La palabra Python aparece:",
      contar_palabra(texto, "Python"), "veces")


# Experimento 2: palabra que no existe
print("\n===== PALABRA NO ENCONTRADA =====")

print("La palabra Java aparece:",
      contar_palabra(texto, "Java"), "veces")


# Experimento 3: repetir una palabra
print("\n===== MÁS REPETICIONES =====")

frase = "hola mundo hola Python hola mundo hola"

print("La palabra hola aparece:",
      contar_palabra(frase, "hola"), "veces")

print("\n===== FIN DEL PROGRAMA =====")
```

**Desafío 43: El Mono que cuenta manzana**

Imagina que tenemos un mono que intenta contar manzanas. Sin embargo, nuestro mono es un poco distraído y olvida cuántos manzanas ha contado cada vez que empieza a contar de nuevo. Cada vez que termina una secuencia de conteo, se olvida de los manzanas contados antes y vuelve a empezar, sumando todos desde el principio. Implementa una función recursiva que simule cómo el mono cuenta manzanas.

Reglas:
- El mono puede contar un manzana a la vez.
- Cada vez que el mono termina de contar una pila de manzanas, tiene que volver a contar desde cero, pero sigue acumulando el total.

Por ejemplo:
- Si el mono tiene 5 manzanas en la pila, contará uno por uno, luego olvida y vuelve a empezar, acumulando la suma.

---

Para más detalles y ejemplos sobre recursividad, puedes consultar la [referencia proporcionada](https://www.tutorialesprogramacionya.com/pythonya/detalleconcepto.php?punto=90&codigo=91&inicio=75#:~:text=En%20Python%20las%20funciones%20o,nuevas%20variables%20locales%20y%20par%C3%A1metros.).

```python

# DESAFÍO 40: MÁXIMO COMÚN DIVISOR CON RECURSIVIDAD

def mcd(a, b):
    # Caso base: cuando b es 0
    if b == 0:
        return abs(a)

    # Caso recursivo
    return mcd(b, a % b)


# Pruebas originales
print("===== CÁLCULO DEL MCD =====")

print("MCD de 8 y 12:", mcd(8, 12))
print("MCD de 48 y 18:", mcd(48, 18))
print("MCD de 0 y 5:", mcd(0, 5))
print("MCD de 17 y 5:", mcd(17, 5))


# Experimento con otros números
print("\n===== OTROS EJEMPLOS =====")

print("MCD de 20 y 30:", mcd(20, 30))
print("MCD de 36 y 60:", mcd(36, 60))
print("MCD de 15 y 15:", mcd(15, 15))

print("\n===== FIN DEL PROGRAMA =====")
```

```python

# DESAFÍO 41: RESALTAR PALABRAS CON RECURSIVIDAD

# Función para resaltar una palabra
def resaltar(palabra):
    return "**" + palabra + "**"


# Función recursiva
def resaltar_palabra(texto, clave):
    palabras = texto.split()

    # Caso base: no quedan palabras
    if len(palabras) == 0:
        return ""

    # Revisamos la primera palabra
    primera = palabras[0]

    if primera.lower() == clave.lower():
        primera = resaltar(primera)

    # Continuamos con el resto del texto
    resto = " ".join(palabras[1:])

    return primera + (" " + resaltar_palabra(resto, clave) if resto else "")


# Texto de ejemplo
texto = "Python es divertido y aprender Python permite crear programas en Python"

print("===== TEXTO ORIGINAL =====")
print(texto)

print("\n===== TEXTO RESALTADO =====")
print(resaltar_palabra(texto, "Python"))


# Experimento con otra palabra
print("\n===== OTRO EJEMPLO =====")

frase = "Aprender a programar requiere practicar y practicar ayuda a aprender"

print(resaltar_palabra(frase, "practicar"))

print("\n===== FIN DEL PROGRAMA =====")
```

```python

# DESAFÍO 42: INVERTIR PALABRAS CON RECURSIVIDAD

def invertir_palabras(oracion):
    # Separamos la oración en palabras
    palabras = oracion.split()
    return invertir_recursivo(palabras)


def invertir_recursivo(palabras):
    # Caso base: una palabra o ninguna
    if len(palabras) <= 1:
        return " ".join(palabras)

    # Caso recursivo
    resto = invertir_recursivo(palabras[1:])
    return resto + " " + palabras[0]


# Pruebas originales
print("===== INVERTIR PALABRAS =====")

print(invertir_palabras("Hola mundo"))
print(invertir_palabras("El perro corre rapido"))
print(invertir_palabras("Hola, mundo!"))
print(invertir_palabras("Que dia tan lindo!"))


# Experimento con otras oraciones
print("\n===== OTROS EJEMPLOS =====")

print(invertir_palabras("Estoy aprendiendo Python"))
print(invertir_palabras("La recursividad es interesante"))
print(invertir_palabras("Python"))

# Prueba con una oración vacía
print("Oración vacía:", repr(invertir_palabras("")))

print("\n===== FIN DEL PROGRAMA =====")
```

```python

# DESAFÍO 43: EL MONO QUE CUENTA MANZANAS

# Función recursiva para contar una pila
def contar_manzanas(manzanas):
    # Caso base: no quedan manzanas
    if manzanas == 0:
        return 0

    # Contamos una manzana y continuamos
    return 1 + contar_manzanas(manzanas - 1)


# Función para repetir el conteo
def mono_distraido(manzanas, rondas):
    # Caso base: no quedan rondas
    if rondas == 0:
        return 0

    # El mono vuelve a contar desde cero
    conteo = contar_manzanas(manzanas)

    print("El mono contó", conteo, "manzanas")

    # Acumulamos el total de las rondas
    return conteo + mono_distraido(manzanas, rondas - 1)


# Ejemplo con 5 manzanas
print("===== MONO QUE CUENTA MANZANAS =====")

total = mono_distraido(5, 3)

print("Total acumulado:", total)


# Experimento con otra pila
print("\n===== OTRO EJEMPLO =====")

total2 = mono_distraido(3, 2)

print("Total acumulado:", total2)

print("\n===== FIN DEL PROGRAMA =====")
```
