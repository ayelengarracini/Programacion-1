Desafío 20: Caracteres ASCII

Eres un desarrollador creando una herramienta educativa que muestra los caracteres ASCII correspondientes a una lista de números. Se recomienda usar la función chr() para convertir los números en caracteres.

```python

# ==========================================
# DESAFÍO 20: CARACTERES ASCII
# CÓDIGO CORREGIDO
# ==========================================

print("===== CONVERSOR DE CARACTERES ASCII =====")

# Lista de números que queremos convertir
numeros = [65, 66, 67, 68, 69, 97, 98, 99]

# Calculamos cuántos números tenemos
cantidad = len(numeros)

print(f"\nCantidad de números: {cantidad}")

print("\n===== RESULTADOS =====")

# Recorremos la lista de números
for i in range(cantidad):

    numero = numeros[i]

    # Verificamos que esté dentro del rango ASCII
    if 0 <= numero <= 127:

        # Convertimos el número a carácter
        caracter = chr(numero)

        print(f"El carácter ASCII de {numero} es: {caracter}")

    else:

        print(f"El número {numero} está fuera del rango ASCII.")

# ==========================================
# EXPERIMENTO: CONVERTIR UNA PALABRA
# ==========================================

print("\n===== EXPERIMENTO EXTRA =====")

# Creamos una palabra utilizando códigos ASCII
codigos = [65, 89, 69, 76, 69, 78]

palabra = ""

for codigo in codigos:
    palabra += chr(codigo)

print(f"La palabra formada es: {palabra}")

print("\n===== FIN DEL PROGRAMA =====")
```
